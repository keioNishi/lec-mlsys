## 画像処理のためのビジュアルプロンプト

```python
import base64

def encode_image(filepath):
    with open(filepath, 'rb') as f:
        image_bytes = f.read()
    encoded = str(base64.b64encode(image_bytes), 'utf-8')
    return "data:image/jpg;base64,"+encoded
```

注意: 下のセルを実行し、マウスを使って画像上に バウンディングボックスを描画すること

```python
IS_COLAB = True
SOURCE_IMAGE_PATH = "dog-3.jpeg"
TARGET_IMAGE_PATH = "dog-8.jpeg"
NAMES = ['cap', 'sunglasses', 'dog']

if IS_COLAB:
    from google.colab import output
    output.enable_custom_widget_manager()

from jupyter_bbox_widget import BBoxWidget

widget = BBoxWidget(classes=NAMES)
widget.image = encode_image(IMAGE_PATH)
widget
```

```python
widget.bboxes
```

**NOTE:** `YOLOEVPSegPredictor` takes `np.ndarray` box prompt in `[x_min, y_min, x_max, y_max]` format.

```python
import numpy as np

default_box = [
    {'x': 216, 'y': 275, 'width': 108, 'height': 83, 'label': 'cap'},
    {'x': 225, 'y': 351, 'width': 84, 'height': 51, 'label': 'sunglasses'},
    {'x': 160, 'y': 383, 'width': 270, 'height': 555, 'label': 'dog'}
]

boxes = widget.bboxes if widget.bboxes else default_box
bboxes = np.array([
    [
        box['x'],
        box['y'],
        box['x'] + box['width'],
        box['y'] + box['height']
    ] for box in boxes
], dtype=np.float64)

cls = np.array([NAMES.index(box['label']) for box in boxes], dtype=np.int32)
```

```python
from ultralytics import YOLOE
from ultralytics.models.yolo.yoloe.predict_vp import YOLOEVPSegPredictor

model = YOLOE("yoloe-v8l-seg.pt").cuda()
prompts = dict(bboxes=bboxes, cls=cls)

source_image = Image.open(SOURCE_IMAGE_PATH)
target_image = Image.open(TARGET_IMAGE_PATH)

model.predict(source_image, prompts=prompts, predictor=YOLOEVPSegPredictor, return_vpe=True)
model.set_classes(NAMES, model.predictor.vpe)
model.predictor = None

results = model.predict(target_image)
detections = sv.Detections.from_ultralytics(results[0])

annotated_image = target_image.copy()
annotated_image = sv.BoxAnnotator().annotate(scene=annotated_image, detections=detections)
annotated_image = sv.LabelAnnotator().annotate(scene=annotated_image, detections=detections)

annotated_image
```

### 画像キャプション

別の画像を読み込んでキャプションを付与する

```python
from huggingface_hub import hf_hub_download
from PIL import Image

def prepare_image():
  filepath = hf_hub_download(repo_id="nielsr/textvqa-sample", filename="bus.png", repo_type="dataset")
  image = Image.open(filepath).convert("RGB")

  return image

image = prepare_image()
image
```

ここでは、Huggingfaceから別のGITモデルとして、COCOデータセットで微調整されたモデルをロードして動作確認する

```python
from transformers import AutoProcessor, AutoModelForCausalLM
import torch

processor = AutoProcessor.from_pretrained("microsoft/git-base-coco")
model = AutoModelForCausalLM.from_pretrained("microsoft/git-base-coco")

model.to(device)
inputs = processor(images=prepare_image(), return_tensors="pt")
pixel_values = inputs.pixel_values.to(device)

generated_ids = model.generate(pixel_values=pixel_values, max_length=50)

print("Generated caption:", processor.batch_decode(generated_ids, skip_special_tokens=True))
```

### ビジュアル質問応答（VQA）

GITは、TextVQAのようなVQAデータセットでも微調整が可能である

バス画像に関連する質問をモデルに投げかけてみる

```python
processor = AutoProcessor.from_pretrained("microsoft/git-base-textvqa")
model = AutoModelForCausalLM.from_pretrained("microsoft/git-base-textvqa")

pixel_values = processor(images=prepare_image(), return_tensors="pt").pixel_values

question = "what does the front of the bus say at the top?"
input_ids = processor(text=question, add_special_tokens=False).input_ids
input_ids = [processor.tokenizer.cls_token_id] + input_ids
input_ids = torch.tensor(input_ids).unsqueeze(0)

generated_ids = model.generate(pixel_values=pixel_values, input_ids=input_ids, max_length=50)

print("Generated answer:", processor.batch_decode(generated_ids[:, input_ids.shape[1]:], skip_special_tokens=True))
```

"what does the front of the bus say at the top?"
つまり、"バスの先頭には何と書いてある？"という質問に対して、"special"と答えており、正解である

GITに送るためには、映像からいくつかのフレームをサンプリングする必要がある

ここでは、1つのビデオから6フレームを抽出して利用している


```python
num_frames = model.config.num_image_with_embedding
print(num_frames)
```

ビデオのデコードには、decordライブラリを使用する
- サンプリングはNumPyを使って行う

```python
import numpy as np
from decord import VideoReader, cpu

def sample_frames(file_path, num_frames):
    # set seed for reproducability
    np.random.seed(45)

    def sample_frame_indices(clip_len, frame_sample_rate, seg_len):
        converted_len = int(clip_len * frame_sample_rate)
        end_idx = np.random.randint(converted_len, seg_len)
        start_idx = end_idx - converted_len
        indices = np.linspace(start_idx, end_idx, num=clip_len)
        indices = np.clip(indices, start_idx, end_idx - 1).astype(np.int64)
        return indices

    # video clip consists of 300 frames (10 seconds at 30 FPS)
    videoreader = VideoReader(file_path, num_threads=1, ctx=cpu(0))

    # sample 6 frames
    videoreader.seek(0)
    indices = sample_frame_indices(clip_len=num_frames, frame_sample_rate=4, seg_len=len(videoreader))
    frames = videoreader.get_batch(indices).asnumpy()

    return list(frames)

frames = sample_frames(file_path, num_frames)
```

モデルの性能は、どのフレームを見るかに大きく依存する

どのフレームをサンプリングしたかを確認する

```python
# visualize first sampled frame
Image.fromarray(frames[0])
```

このフレームを用いてフォワードパスを実行する

```python
# move model to GPU
model.to(device)
# prepare frames for the model
inputs = processor(images=frames, return_tensors="pt").to(device)

generated_ids = model.generate(pixel_values=inputs.pixel_values, max_length=50)

print("Generated caption:", processor.batch_decode(generated_ids, skip_special_tokens=True))
```

```python
# GIT fine-tuned on a video QA dataset
processor = AutoProcessor.from_pretrained("microsoft/git-base-msrvtt-qa")
model = AutoModelForCausalLM.from_pretrained("microsoft/git-base-msrvtt-qa")
```

```python
# prepare video
inputs = processor(images=frames, return_tensors="pt").to(device)

# prepare question
question = "what is the woman eating?"

input_ids = processor.tokenizer.encode(text=question, add_special_tokens=False)
input_ids = [processor.tokenizer.cls_token_id] + input_ids
input_ids = torch.tensor(input_ids).unsqueeze(0)

print("Generating...")
model.to(device)
input_ids = input_ids.to(device)
generated_ids = model.generate(pixel_values=inputs.pixel_values, input_ids=input_ids, max_length=50)

print("Generated:", processor.batch_decode(generated_ids, skip_special_tokens=True))
```

## LibriSpeechデータセットのロード

torchaudioを使って、LibriSpeechコーパスのtest-clean分割をロードする

```python
import os
import numpy as np

try:
    import tensorflow  # required in Colab to avoid protobuf compatibility issues
except ImportError:
    pass

import torch
import pandas as pd
import whisper
import torchaudio

from tqdm.notebook import tqdm


DEVICE = "cuda" if torch.cuda.is_available() else "cpu"
```

```python
class LibriSpeech(torch.utils.data.Dataset):
    """
    A simple class to wrap LibriSpeech and trim/pad the audio to 30 seconds.
    It will drop the last few seconds of a very small portion of the utterances.
    """
    def __init__(self, split="test-clean", device=DEVICE):
        self.dataset = torchaudio.datasets.LIBRISPEECH(
            root=os.path.expanduser("~/.cache"),
            url=split,
            download=True,
        )
        self.device = device

    def __len__(self):
        return len(self.dataset)

    def __getitem__(self, item):
        audio, sample_rate, text, _, _, _ = self.dataset[item]
        assert sample_rate == 16000
        audio = whisper.pad_or_trim(audio.flatten()).to(self.device)
        mel = whisper.log_mel_spectrogram(audio)

        return (mel, text)
```

```python
dataset = LibriSpeech("test-clean")
loader = torch.utils.data.DataLoader(dataset, batch_size=16)
```

## Whisperモデルを使ったデータセットの推論の実行

データセットの全発言を書き起こす

```python
model = whisper.load_model("base.en")
print(
    f"Model is {'multilingual' if model.is_multilingual else 'English-only'} "
    f"and has {sum(np.prod(p.shape) for p in model.parameters()):,} parameters."
)
```

 短文の文字起こしをタイムスタンプなしで行う

```python
options = whisper.DecodingOptions(language="en", without_timestamps=True)
```

```python
!pip install torchcodec
hypotheses = []
references = []

for mels, texts in tqdm(loader):
    results = model.decode(mels, options)
    hypotheses.extend([result.text for result in results])
    references.extend(texts)
```

```python
data = pd.DataFrame(dict(hypothesis=hypotheses, reference=references))
data
```

## 単語誤り率の計算

英語ノーマライザーの実装を使用して書き起こしを標準化し、WERを計算する

```python
import jiwer
from whisper.normalizers import EnglishTextNormalizer

normalizer = EnglishTextNormalizer()
```

```python
data["hypothesis_clean"] = [normalizer(text) for text in data["hypothesis"]]
data["reference_clean"] = [normalizer(text) for text in data["reference"]]
data
```

```python
wer = jiwer.wer(list(data["reference_clean"]), list(data["hypothesis_clean"]))

print(f"WER: {wer * 100:.2f} %")
```

## 日本語の文字起こし

では、早速完全機能を用いて実験する

文字起こしを実行したいファイルが必要であるが、
対応しているフォーマットは比較的多いようである
- ここでは、mp3とm4aを用いて実験する

ナレーション音声のサンプルを用いる
- 発音がきれいであるため、是非皆さんの地声で実験してほしい
- ノイズ環境でも動作することを確認するとよい


```python
import os
import locale
locale.getpreferredencoding = lambda: "UTF-8"

# Install gTTS to generate sample audio
!pip install -q gTTS

from gtts import gTTS

print("Generating sample audio files...")

# Generate sample1.mp3 (Japanese)
text_ja = "これはウィスパーの音声認識テストです。正しく文字起こしができるか確認します。"
tts_ja = gTTS(text=text_ja, lang='ja')
tts_ja.save("sample1.mp3")

# Generate sample1.m4a (Convert mp3 to m4a using ffmpeg)
!ffmpeg -hide_banner -loglevel error -i sample1.mp3 -c:a aac -b:a 128k sample1.m4a -y

# Generate sample2.m4a (English for language detection)
text_en = "This is a test for language detection using Whisper."
tts_en = gTTS(text=text_en, lang='en')
tts_en.save("sample2.temp.mp3")
!ffmpeg -hide_banner -loglevel error -i sample2.temp.mp3 -c:a aac -b:a 128k sample2.m4a -y

# Generate sample3.m4a (Duplicate for noise robustness test)
# Note: Ideally this would have noise, but we use clean audio to ensure execution
!cp sample1.m4a sample3.m4a

print("Done. Created: sample1.mp3, sample1.m4a, sample2.m4a, sample3.m4a")
```

まず、普通の方法で文字化してみる

```python
model = whisper.load_model("base")
result = model.transcribe("sample1.mp3")
print(result["text"])
```

mp3の他、様々なフォーマットに対応している
- 若干マイナーなm4aを用いてみる

```python
result = model.transcribe("sample1.m4a")
print(result["text"])
```

別の形で変換する
- 処理を分解することで、言語推定も可能となる

```python
audio = whisper.load_audio("sample2.m4a")
audio = whisper.pad_or_trim(audio)
mel = whisper.log_mel_spectrogram(audio).to(model.device)
# detect the spoken language
_, probs = model.detect_language(mel)
print(f"Detected language: {max(probs, key=probs.get)}")
# decode the audio
options = whisper.DecodingOptions()
result = whisper.decode(model, mel, options)
# print the recognized text
print(result.text)
```

雑音があってもロバストである

```python
result = model.transcribe("sample3.m4a")
print(result["text"])
```
