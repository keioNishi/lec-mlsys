## 問1  LLMとは、Transformerのことか

**答え**

違う

LLM（Large Language Model：大規模言語モデル）は、「大量の文章で学習した、大きな言語モデル」という**用途と規模**を表す言葉で、中の構造を決めていない 今の代表的なLLMはTransformerという構造を使っているが、Transformerでない部分を含むLLMもある

日常では、検索やファイル処理や画面まで含んだ**サービス全体**を「LLM」と呼ぶことが多い 本テキストでは、学習済みのモデル本体と、それを使うシステムを、言葉として区別する

**説明**

LLMという言葉には、2つの使い方がある

| 使い方 | 指しているもの | 例 |
| --- | --- | --- |
| 狭い意味 | 学習済みの言語モデルそのもの（重みの集まり） | `Qwen3-4B` のモデルファイル |
| 広い意味 | モデルを中心にした対話サービス全体 | ChatGPTのような製品（検索・計算・画面を含む） |

「このLLMはWeb検索ができた」からといって、「TransformerやTransformerのDecoder自体が通信した」わけではない
- Transformerについての構造は既に学んだ通りで、どこにもWebにアクセスする仕組みはない
- モデルが「検索してほしい」という要求を文字で出し、外側のプログラムがネットワークにアクセスした、という可能性を分けて考える必要がある（問18で扱う）

 例えば Jamba は、TransformerとMamba（状態空間モデル）を組み合わせたLLMである（Lieber et al., 2024）
- Transformerを使うモデルでも、入力の分割（トークン化）、次のトークンの選択、生成の停止、検索、プログラムの実行までが全部Transformerというわけではない

以下、学習済みではあるが、今一度まとめておく
- もとのTransformer（Vaswani et al., 2017）は翻訳のためのモデルで、入力の文を読む**エンコーダ**と、出力の文を1語ずつ作る**デコーダ**の2つからなる
- GPTに始まる今のLLMの大半は、このうち**デコーダだけ**を使う（Decoder-only） エンコーダが無いので、エンコーダの出力を参照するAttention（Cross-Attention）も無い
- デコーダの特徴は、**自分より前の位置しか参照しない**こと（因果マスク）である これが、「1つずつ順に生成する」ことと対応している（問3、問6）

| エンコーダ＋デコーダ | もとのTransformer、T5 | 翻訳、要約 |

## 問2  数や問題は無限にあるのに、有限のモデルで扱えるのか

**答え**

扱える
- ただし「規則が有限だから計算できる」と考えると、半分しか合っていない

- 足し算の規則は短い 「桁を足して、10を超えたら繰り上げる」だけで、どんな桁数の数にも使える 規則を覚えるのに、無限の記憶はいらない
- しかし、その規則を**何回も適用する**必要はなくならない 100桁なら100回である この「実行の手間」はどこかで払う必要がある

「その手間を、LLMはどこで払っているのか」を調べてみよう

**説明**

疑問を順番に並べると、次のようになる

| 段階 | 疑問 | 区別すべきこと |
| --- | --- | --- |
| 出発点 | 数や問題は無限にある 有限のモデルで扱えるのか | 入力の種類の数と、規則の長さ |
| 次の疑問 | 規則が有限でも、実行には大量の手順がいるのではないか | 規則を覚えることと、実行の時間・作業の場所 |
| 核心 | 例えば80層のモデルで、長い手順をどこで実行するのか | 1回の計算と、生成の繰り返し（問6） |
| 見かけの矛盾 | 途中を表示せず、「答えだけ」でも答えられるのはなぜか | 表示の長さと、内部で実行した計算量（問12） |
| 広がり | 計算機や検索を外側から使っているのではないか | モデル本体と、モデルを使うシステム（問18） |

まず、足し算の規則を式で書くと、数字を $a_i, b_i$、下の桁からの繰り上がりを $c_i$ として、

$$
s_i=(a_i+b_i+c_i)\bmod 10,
\qquad
c_{i+1}=\left\lfloor\frac{a_i+b_i+c_i}{10}\right\rfloor
$$

である

この2行を、桁の数だけ繰り返す 繰り上がり $c_i$ は次の桁の入力になるので、**前の桁が終わるまで次の桁に進めない**（これを「逐次の依存」と呼ぶ）

計算に必要な「資源」を4つに分けて考えると、見通しがよい

| 資源 | 問い | 例 |
| --- | --- | --- |
| 表現・記憶の容量 | 規則や状態を、どれだけ区別して保てるか | 重みの数、精度、文脈の長さ |
| 演算の総量 | 全部でどれだけ計算するか | 行列の積、外部プログラムの演算 |
| 逐次の深さ | 前の結果を待つ段が、どれだけ続くか | モデルの層、生成の繰り返し、ツールの待ち |
| 作業記憶 | 途中の状態をどこに置くか | 生成した文字、KVキャッシュ、外部のファイル |

「規則を小さく表現できる」は、1番目の話である 2〜4番目の費用が消えるわけではない
- 「無限の組み合わせ」は自然言語にもあり、言葉は有限、計算は無限」というわけでもないが、計算はより分かりやすいであろう
- 大事なのは、対象に共通の構造があるか、その構造をモデルが学習できたか、正解の判定が厳密か、である

### 確かめる ― 有限の規則を、桁の数だけ繰り返す

何を実装しているか

- 上の2つの式を、そのままPythonの関数にする 規則は2行だが、適用する回数は桁数に比例する
- 乱数で作った最大200桁の整数の和を、Pythonの整数演算と比べる

実装の肝

- 繰り上がり $c_i$ が次の桁の入力になる つまり、桁の数だけ「待ち」が続く
- 結果が一致しなければ `assert` で止まる

結果の読み方

- 規則の長さ（関数の行数）は変わらないが、必要な規則の適用回数は、桁数と一緒に増える これが「規則を覚えること」と「実行の手間」の違いである

観測上の限界

- 筆算の規則に待ちがあるからといって、**もっと短い別の方法が存在しない**ことは示せない 繰り上がりを並列に予測する回路もある（問4）

````python
# 桁ごとの加算規則（有限の規則）を実装し、1桁ずつの動きと、桁数ごとの適用回数を見る
def add_by_digits(x, y, show=False):
    """筆算と同じ規則で x + y を求める  返り値は（和、規則を適用した回数）"""
    a, b = str(x)[::-1], str(y)[::-1]            # 下の桁から処理するので、文字列を逆順にする
    carry, digits, steps = 0, [], 0
    for i in range(max(len(a), len(b))):
        ai = int(a[i]) if i < len(a) else 0      # 桁が足りなければ 0 とみなす
        bi = int(b[i]) if i < len(b) else 0
        s = ai + bi + carry
        if show:
            print(f"  {i + 1}桁目: {ai} + {bi} + 繰り上がり{carry} = {s:2d} → この桁は {s % 10}、次への繰り上がりは {s // 10}")
        digits.append(s % 10)                    # s_i = (a_i + b_i + c_i) mod 10
        carry = s // 10                          # c_{i+1} = floor((a_i + b_i + c_i) / 10)
        steps += 1
    if carry:
        digits.append(carry)                     # 最後に残った繰り上がりが、最上位の桁になる
    return int("".join(map(str, digits[::-1]))), steps


# (1) 小さな例で、1桁ずつの動きを見る
x, y = 478, 256
print(f"{x} + {y} を、下の桁から順に計算する")
z, steps = add_by_digits(x, y, show=True)
print(f"  答え {z}（規則を {steps} 回適用）  Pythonの足し算 {x + y} と一致: {z == x + y}")

# (2) 桁数を増やすと、規則の適用回数がどう増えるか
print("\n桁数   規則の適用回数   Pythonの足し算と一致")
rng = random.Random(1)
for n_digits in (2, 8, 32, 100, 200):
    x = rng.randrange(10 ** (n_digits - 1), 10 ** n_digits)
    y = rng.randrange(10 ** (n_digits - 1), 10 ** n_digits)
    z, steps = add_by_digits(x, y)
    assert z == x + y
    print(f"{n_digits:4d}   {steps:8d}         {z == x + y}")
print("\n規則（関数の中身）は同じまま、適用回数だけが桁数に比例して増える")
````

## 問3  Transformerの1回の処理では、何が起きているのか

**答え**

入力の文章をトークン（単語や文字の断片）に分け、**トークンごとに1本のベクトル**を持たせる そのベクトルを、L個の層で順に変換し、最後に「次に来るトークンの確率」を出す

ポイントは2つ

- 文章全体を1本のベクトルに押し込むのではない トークンの数だけベクトルがあり、互いに参照し合う（Attention）
- 「次のトークンを予測するだけ」は、**出力の形**の説明である 中では多数の行列の計算が行われている

**説明**

この内容も既習済であるが、入力の長さを $n$、ベクトルの次元を $d$ とすると、ある層の状態は $n$ 本の $d$ 次元ベクトル、つまり

$$
H^{(\ell)}\in\mathbb{R}^{n\times d}
$$

である（Vaswani et al., 2017） 例えば100桁の数字は、「100桁の整数専用の1点」として覚える必要はない 数字の列と、その位置関係として扱える ただし、数字の区切り方（1桁が1トークンか）はモデルによって違う

<img src="https://class.west.sd.keio.ac.jp/dataai/text/llmu-02-block.png" width=760>

上の図は、1つのブロックの中身である 2つの部品がある

- **Attention（注意機構）**: 他の位置の情報を集める
  - 1ヘッドの式は $\operatorname{softmax}(QK^{\mathsf T}/\sqrt{d_k}+M)V$ で、$Q$ は「何を探すか」、$K$ は「照合用の札」、$V$ は「取り出す中身」である
  - $M$ は、まだ生成していない未来のトークンを見ないようにするマスクである
- **FFN（Feed-Forward Network）**: 集めた情報を、位置ごとに非線形に変換する
  - 「Attentionが読む部分、FFNが加工する部分」という説明は目安であり、厳密な役割分担ではない
  - ある機能が複数のヘッドや層に分散して実現されることもある

最後の位置のベクトル $h$ を出力ヘッドで語彙ごとのスコア $z$ に変え、確率にする

$$
z=W_{\mathrm{out}}h+b,
\qquad
p=\operatorname{softmax}(z)
$$

ここまでが「1回の前向き計算」である
- その後、生成エンジンが確率の高い候補を選ぶ（または確率に従って抽選する） ここで初めて次のトークンが決まる

### 確かめる ― 1回の前向き計算で何が得られるか

何を実装しているか

- 式 `24×17+38×26−75=` を入力し、**1回の前向き計算**で、全位置の表現と、全位置の「次のトークンの分布」が得られることを確かめる

実装の肝

- `output_hidden_states=True` で、埋め込み（第0層）から最終層までの表現 $H^{(0)},\dots,H^{(L)}$ を取り出す
- 表現の形は `(1, n, d)`、出力の形は `(1, n, 語彙数)` である 最後の位置の分布だけが「次のトークン」に使われる
- 形を `assert` で検査する 構造の違うモデルに差し替えると、ここで止まる

結果の読み方

- 入力が `n` トークンでも、前向き計算は1回である 位置は並列に処理される
- 「1回の前向き計算」は1つの算術命令ではなく、$L$ 層ぶんの行列演算を含む計算全体である

観測上の限界

- 次のトークンの分布が得られることは、その分布が**正しい**ことを意味しない

- ここで語彙数が151,936と出るが、これは、モデルの埋め込み行列と出力ヘッドの行数である
  - トークナイザが持っている基本の語彙はこれよりも少なく、141,643、さらに特殊トークンが26個あり、これを加えても少し少ない

- モデル規模が小さすぎるため、演算は正しく行えていない
  - いきなり、?を予測している

````python
# 1回の前向き計算で得られるものを、形と中身で確かめる
def show_next_token(prompt, k=5):
    enc = tok_s(prompt, return_tensors="pt").to(device)
    n = enc.input_ids.shape[1]
    with torch.no_grad():
        out = small(**enc, output_hidden_states=True)          # 前向き計算は、この1回だけ
    L, d, V = cfg.num_hidden_layers, cfg.hidden_size, cfg.vocab_size
    assert len(out.hidden_states) == L + 1
    assert out.hidden_states[-1].shape == (1, n, d)
    assert out.logits.shape == (1, n, V)
    print(f"入力: {prompt!r}")
    print(f"  トークンに分けると {n} 個: {tok_s.convert_ids_to_tokens(enc.input_ids[0])}")
    print(f"  前向き計算1回で得られるもの: 各位置のベクトル {n} 本 × {d} 次元、各位置の「次のトークンの確率」 {n} 組 × 語彙 {V} 個")
    p = torch.softmax(out.logits[0, -1], dim=-1)               # 最後の位置の確率だけが、次のトークンの候補になる
    top = torch.topk(p, k)
    print("  次に来るトークンの候補（確率の高い順）")
    for rank, (prob, idx) in enumerate(zip(top.values.tolist(), top.indices.tolist()), 1):
        text = tok_s.decode(idx).replace("\n", "⏎").replace(" ", "␣")   # 改行と空白を、見える記号に置き換える
        print(f"    {rank}位  「{text}」  確率 {prob:.3f}")
    print()


show_next_token("The capital of France is")
show_next_token("24×17+38×26−75=")
````

### 確かめる ― 層を進むと、予測はどう変わるか（ロジットレンズ）

何を実装しているか

- 途中の層のベクトルを、最終層と同じ手順（正規化と出力ヘッド）で無理やり語彙に読み替え、「その層で止めて答えさせたら何を出すか」を層ごとに見る
  - この手法をロジットレンズ（logit lens）と呼ぶ

実装の肝

- 最終層の値（`hidden_states[-1]`）は、すでに最終の正規化を通っている
  - 全層に同じ処理を当てると最終層だけ二重に正規化するので、分岐している
  - 最終層の読み替えが通常の出力と一致することを `assert` で確かめる

結果の読み方

- 4層ごとに、その層の予測を並べる
 - **どの層から、正しい答えが現れるか**を見る

観測上の限界

- 途中の層のベクトルは、最終層で読まれることを前提にした作業用の特徴である
  - 読めないことは、情報が無いことを意味しない

最後まで達成しないと答えが出ていないわけではない点に注意する

````python
# ロジットレンズ: 各層の表現を、出力ヘッドで直接トークンに読み替える
def logit_lens(prompt):
    enc = tok_s(prompt, return_tensors="pt").to(device)
    with torch.no_grad():
        out = small(**enc, output_hidden_states=True)
        rows = []
        for layer, h in enumerate(out.hidden_states):
            x = h[0, -1]
            if layer < len(out.hidden_states) - 1:      # 最終層の値は、すでに最終の正規化を通っている
                x = small.model.norm(x)
            rows.append(tok_s.decode(small.lm_head(x).argmax()))
    # 実装の検査: 最終層の読み替えが、通常の出力と一致する
    assert rows[-1] == tok_s.decode(out.logits[0, -1].argmax())
    return rows


for prompt in ["The capital of France is", "The opposite of hot is"]:
    rows = logit_lens(prompt)
    print(prompt)
    print("  " + " | ".join(f"層{l}:{rows[l].strip() or '␣'}" for l in range(0, len(rows), 4)) + f" | 層{len(rows) - 1}:{rows[-1].strip()}")
````

結果の読み方

- `The capital of France is` では、浅い層では答えと無関係なトークン（`not`、`a`、`____` など）が並び、層24あたりから ` Paris` に切り替わる `The opposite of hot is` でも同じで、終盤の層で ` cold` になる
- 答えは1つの層で「検索」されるのではなく、**層を重ねるあいだに少しずつ組み立てられる**
- 中間層のトークンは人間にとって意味を持たない 途中の表現は、人間が読める「思考」ではない（問14で再び扱う）
- 数式 `24×17+38×26−75=` に対しては、この小さなモデルは答えを出せない 小型モデルが、桁の多い計算を1回の前向き計算で解けるとは限らない

## 問4  80層のモデルは、80回しか計算できないのか

**答え**

違う

1つの層は「1回の足し算」ではなく、多数の位置・次元・ヘッドにわたる大量の行列演算を含む
- だから「80層だから80回」ではない

ただし逆に、層の中に演算が大量にあっても、**前の結果を待たなければ進めない計算**を、無限に片づけられるわけでもない
- ここを理解するために、「計算の総量」と「待ちの段数（深さ）」を分けて数える

**説明**

<img src="https://class.west.sd.keio.ac.jp/dataai/text/llmu-03-work-depth.png" width=760>

上の図は、8個の数の合計を2通りで求めたものである どちらも足し算は7回だが、左は前の結果を待つ段が7段、右は3段である

1024個の数を左から順に足すと1023段の待ちが生じる
- 2個ずつ組にして足す木構造なら、足し算の回数は1023回のまま、段数は10段になる
- これは足し算の結合則から言える

「プログラムが100万回ループする」ことと、「どんな方法でも100万段の待ちが避けられない」ことは違う
- 別の並列な方法や、数式の簡略化があり得る
- ただし、同じ並べ替えを浮動小数点で行うと、丸めの順序が変わって結果が一致しないことがある
  - 並べ替えてよいかは、演算の意味による
- **モデルの層数、回路の論理深さ、機械命令の数、実時間は、同じ単位ではない**

本当に短縮できない長い待ちが、モデルの層数で処理できる範囲を超えるなら、生成の繰り返し（問6）、内部の再帰構造（問34）、外部プログラム（問18）のどれかが必要になる
  - 「ベクトルの次元を大きくした」という説明だけでは、実行を説明したことにならない

### 確かめる ― 計算の総量と、待ちの段数を数える

何を実装しているか

- 1024個の数の合計を、(a) 左から順に、(b) 2個ずつ組にして、の2通りで求め、**足し算の回数**と**待ちの段数**を数える
- 同じ並べ替えを float32 で行い、結果が一致しないことを確かめる

実装の肝

- (b) では、各段で隣同士を足して要素数を半分にする 段数は $\lceil\log_2 1024\rceil=10$ である
- 整数では (a) と (b) が必ず一致するので `assert` で確かめる float32 では、大きさの違う値を混ぜると一致しなくなることを `assert` で確かめ、厳密な値との誤差も表示する

結果の読み方

- 足し算の回数はどちらも1023回で、変わるのは段数だけである
- float32 では、足す順序で結果が変わる 並列化が算術の意味を変え得る例である

観測上の限界

- どの計算も木の形に並べ替えられるわけではない 任意の長い待ちを、常に浅くできるわけではない

````python
# 総手順数（加算の回数）と、逐次深さ（待ち合わせの段数）を数える
def sum_sequential(xs):
    total, adds = xs[0], 0
    for v in xs[1:]:
        total, adds = total + v, adds + 1
    return total, adds, adds                          # 深さ = 加算回数


def sum_tree(xs):
    adds, depth = 0, 0
    while len(xs) > 1:
        nxt = [xs[i] + xs[i + 1] for i in range(0, len(xs) - 1, 2)]
        adds += len(nxt)
        if len(xs) % 2:
            nxt.append(xs[-1])
        xs, depth = nxt, depth + 1
    return xs[0], adds, depth


ints = [random.randrange(10 ** 6) for _ in range(1024)]
(s1, a1, d1), (s2, a2, d2) = sum_sequential(ints), sum_tree(ints)
assert s1 == s2 == sum(ints) and a1 == a2 == 1023 and d1 == 1023 and d2 == 10
print(f"整数: 逐次 加算{a1}回 深さ{d1} / 木 加算{a2}回 深さ{d2} / 結果は一致")

rng_f = np.random.default_rng(0)                         # 大きさの違う値（10^-3 〜 10^4 程度）を混ぜる
floats = (rng_f.normal(size=1024) * 10.0 ** rng_f.integers(-3, 5, size=1024)).astype(np.float32)
f1, _, _ = sum_sequential([np.float32(v) for v in floats])
f2, _, _ = sum_tree([np.float32(v) for v in floats])
exact = math.fsum(float(v) for v in floats)                # 丸めなしの厳密な和（倍精度で正確に足し合わせる）
assert f1 != f2                                            # 足す順序を変えると、float32の結果は一致しない
print(f"float32: 逐次 {float(f1):.6f} / 木 {float(f2):.6f} / 厳密 {exact:.6f}")
print(f"  逐次と木の差: {abs(float(f1) - float(f2)):.2e}（足す順序で結果が変わる）")
print(f"  厳密な値との誤差: 逐次 {abs(float(f1) - exact):.2e} / 木 {abs(float(f2) - exact):.2e}")
````

## 問5  コンピュータの上で動くのに、なぜ計算を間違えるのか

**答え**

モデルを動かすための計算と、利用者の式を計算することは、別物だから

- コンピュータは、モデルの行列演算を正確に実行している
- しかし、入力の `24 × 17` は、プログラムの式として評価されるのではない 文字の列として埋め込まれ、学習した重みで変換されて、「それらしい次の文字」が出てくる `24` のベクトルは、整数24をレジスタに入れたものではない

つまり、**ネットワークを実行するための算術**と、**利用者の数式を正しく評価する算術**は違う
  - 後者を確実にしたければ、計算プログラムにつなぐ（問18、問22）

**説明**

では、モデルの中に「計算の回路」は無いのか
- あることはある
- 加算だけを学習させた小さなTransformerを解析すると、桁に対応した処理が形成されていることが報告されている（Quirke and Barez, 2023）
- 一方、一般用途のモデルの算術を解析した研究では、1つの正確な筆算アルゴリズムではなく、「数値の範囲に反応する複数のヒューリスティック」の寄せ集めで答えていることが報告されている（Nikankin et al., 2024）

ここから言えるのは、**計算に関係する内部回路は形成され得るが、すべてのモデルが正確なアルゴリズムを持つとは限らない**ということである
- 「内部回路」は独立した電卓モジュールではなく、重み・ヘッド・ニューロンの組み合わせが、入力に応じて計算に関係する変換を行う、という意味である

正答した事実から「厳密な計算法を実行した」とは言えないし、「予測しているだけだから計算していない」とも言えない
- どちらも一括には結論できない（問41）

## 問6  層の数が少ないのに、長い手順の計算をどこで実行しているのか

**答え**

**同じモデルを、時間方向に何度も使っている**

- LLMは、次のトークンを1つ選んでは入力の末尾に足し、もう一度モデル全体を通す、という繰り返しで文章を作る
- 1000トークン生成すれば、80層のモデルは80層×1000回、延べ8万回のブロック計算をしたことになる

途中の計算結果（例えば `24 × 17 = 408`）を文字として出せば、それは次の入力になる
- こうして、層の数では足りない長い手順を、**生成の回数**で補える

**説明**

問題文を $x$、これまでに生成したトークンを $y_1,\ldots,y_{t-1}$、学習済みの重みを $\theta$ とすると、次のトークンは

$$
y_t\sim p_\theta(\,\cdot\mid x,y_1,\ldots,y_{t-1})
$$

から選ばれる（Brown et al., 2020） 大事なのは、**$\theta$ は変わらず、入力と作業状態だけが変わる**ことである

<img src="https://class.west.sd.keio.ac.jp/dataai/text/llmu-04-prefill-decode.png" width=760>

上の図のように、生成は2段階に分かれる

- **Prefill（入力の処理）**: 入力のトークン列をまとめて1回の前向き計算で処理する 位置は並列に扱える 最後の位置の出力から、最初の生成トークンの分布が得られる
- **Decode（逐次生成）**: 選んだトークンを1つ取り込み、次の分布を得る 過去の位置の計算結果は、KVキャッシュ（問8）で再利用する

ここでいう「並列」は、同じ層の中で位置をまとめて処理できる、という意味である 層と層の依存や、まだ決まっていない次のトークンへの依存は消えない

<img src="https://class.west.sd.keio.ac.jp/dataai/text/llmu-05-unroll.png" width=760>

上の図は、同じ重みのモデルを時刻ごとに繰り返し使う様子である
- 列が増えても、重みが列の数だけ増えるわけではない
- 「80層×1000トークン≈8万回」という数え方は、**固定された層数と、生成全体で使える時間方向の計算量は別である**ことを示す
- ただし、8万手順の任意のプログラムを正しく実行した、という意味ではない
- 投機的デコーディング（問36）のように、1回の前向き計算で複数トークンを確定する方式もある

### 確かめる ― 生成ループを自分で書き、`generate` と突き合わせる

何を実装しているか

- Prefill 1回 ＋ Decode の繰り返し、という生成ループを、`model(...)` の呼び出しだけで自作する
- 同じ入力から、(1) KVキャッシュを使う自作ループ、(2) キャッシュを使わず毎回全体を計算し直すループ、(3) ライブラリの `generate`、の3つで64トークンを生成し、**トークン列が一致する**ことを確かめる

実装の肝

- 自作ループの前向き計算の回数は、生成トークン数と同じ64回である（Prefillの1回が最初のトークンの分布を作るので、Decodeは63回）
- `generate` の呼び出しは1回でも、内部では前向き計算が64回起きている これが問7の「1回」の違いである
- 一致の比較は float32 の小型モデルで行う 低精度（float16、bfloat16）では、丸めの順序が変わって一致しないことがある

結果の読み方

- 3つのトークン列が一致すれば、「`generate` の中身は、この自作ループと本質的に同じ」「KVキャッシュは計算の再利用であって、別の計算ではない」ことの確認になる
- 所要時間も表示する 入力が短いとキャッシュの有無で差は出ない（次のセルで、長い入力を試す）

観測上の限界

- 一致は float32・貪欲法という条件での観測である ビット単位の一致を一般に保証するものではない

````python
# 生成ループを自作する: Prefill 1回 + Decode を繰り返す（KVキャッシュあり）
def generate_with_cache(model, ids, n_new):
    calls = 0
    with torch.no_grad():
        out = model(input_ids=ids, use_cache=True)                  # Prefill: 入力全体を1回で処理する
        calls += 1
        cache, generated = out.past_key_values, []
        nxt = out.logits[0, -1].argmax().item()
        generated.append(nxt)
        for _ in range(n_new - 1):                                  # Decode: 1トークンずつ取り込む
            out = model(input_ids=torch.tensor([[nxt]], device=device), past_key_values=cache, use_cache=True)
            calls += 1
            cache, nxt = out.past_key_values, out.logits[0, -1].argmax().item()
            generated.append(nxt)
    return generated, calls, cache


def generate_without_cache(model, ids, n_new):
    """キャッシュを使わず、毎回、入力＋生成済みの全体を最初から計算し直す"""
    cur, generated = ids, []
    with torch.no_grad():
        for _ in range(n_new):
            nxt = model(input_ids=cur, use_cache=False).logits[0, -1].argmax().item()
            generated.append(nxt)
            cur = torch.cat([cur, torch.tensor([[nxt]], device=device)], dim=1)
    return generated


def timed(fn, *args):
    torch.cuda.synchronize()
    t0 = time.time()
    result = fn(*args)
    torch.cuda.synchronize()
    return result, time.time() - t0


N_NEW = 64
ids = tok_s("The capital of France is", return_tensors="pt").input_ids.to(device)
(g_cache, calls, cache), t_cache = timed(generate_with_cache, small, ids, N_NEW)
g_nocache, t_nocache = timed(generate_without_cache, small, ids, N_NEW)
with torch.no_grad():
    g_api = small.generate(ids, max_new_tokens=N_NEW, min_new_tokens=N_NEW, do_sample=False)[0, ids.shape[1]:].tolist()

assert calls == N_NEW
assert g_cache == g_api == g_nocache
print(f"前向き計算の回数: {calls}（生成トークン数 {N_NEW} と同じ）")
print(f"3つの生成列が一致: キャッシュあり={g_cache == g_api}、キャッシュなし={g_nocache == g_api}")
print(f"所要時間: キャッシュあり {t_cache:.2f}秒 / キャッシュなし {t_nocache:.2f}秒（入力が短いので差は小さい）")
print("生成された文:", tok_s.decode(g_cache)[:120].replace("\n", " "))
````

### 確かめる ― 入力が長いときのキャッシュの効果

何を実装しているか

- 同じ2つのループを約1400トークンの入力で動かし、32トークンを生成する時間を比べる

実装の肝

- キャッシュなしは、1トークン出すたびに入力全体を計算し直す 入力が長いほど1ステップの費用が増える
- キャッシュありは、Prefillで入力を1回処理したあと、新しい1トークンぶんだけを計算する

結果の読み方

- 「なしは あり の何倍か」を読む 生成列が一致すれば、ここでもキャッシュは答えを変えていない

観測上の限界

- 小型モデルで1回だけ測った値である GPUの状態で時間は変わる

````python
# 入力が長いとき、キャッシュの有無で所要時間がどう変わるか（同じ小型モデル）
long_text = " ".join(["The quick brown fox jumps over the lazy dog."] * 140)     # 約1400トークン
ids_long = tok_s(long_text, return_tensors="pt").input_ids.to(device)
print(f"入力の長さ: {ids_long.shape[1]} トークン")
(g1, _, _), t1 = timed(generate_with_cache, small, ids_long, 32)
g2, t2 = timed(generate_without_cache, small, ids_long, 32)
print(f"キャッシュあり {t1:.2f}秒 / キャッシュなし {t2:.2f}秒 → なしは あり の {t2 / t1:.1f} 倍")
print("生成列の一致:", g1 == g2)
````

結果の読み方

- 短い入力（5トークン、64トークン生成）では、キャッシュの効果は時間に現れない 実行例では、キャッシュありが12.84秒、なしが8.33秒で、ありのほうが長い
  - 小型モデルで入力が短いと、1回の呼び出しの固定費が支配的だからである
  - 先に実行するキャッシュありの側には、GPUでの最初の計算の準備にかかる時間も加わる この2つの数字の差を、優劣として読まない
- 約1400トークンの入力では、キャッシュなしは毎回1400トークンぶんを計算し直すので、ありの6.3倍かかる（実行例では、ありが4.02秒、なしが25.45秒）
  - この倍率は、GPUが速いほど小さく出る 固定費の割合が大きくなるためである
- どちらの条件でも生成列は一致する **キャッシュは答えを変えず、再計算を省く仕組みである**

### 新しいトークンを足しても、なぜ全部を計算し直さなくてよいのか

「Attentionは全体を参照するのだから、新しいトークンを1つ足せば、次の層からは影響が全体に波及し、結局すべて計算し直しになるのではないか」という疑問が出る

そうはならない **新しいトークンを足しても、過去の位置の表現は、どの層でも変わらない** よって、計算し直す必要がない

なぜ波及しないのか

- デコーダのAttentionには因果マスクがあり、各位置は自分より前の位置しか参照できない 情報の流れは一方通行である
  - 過去 → 新しいトークン: 流れる（新しい位置は、過去の全位置を参照する）
  - 新しいトークン → 過去: 流れない（過去の位置は、後ろを見ない）

<img src="https://class.west.sd.keio.ac.jp/dataai/text/llmu-14-direction.png" width=760>

- 位置3の第2層の表現は、位置1〜3の第1層の表現だけから決まる 位置6が後から増えても、位置3の計算には一切入ってこない
- これが全層で成り立つので、過去の位置は第1層から最終層まで、すべて前回の計算結果のまま使える
- 「次の層からは全体に波及する」という理解は、エンコーダ（BERTなど）なら正しい エンコーダは前後の両方を参照するので、1語足すと全位置が変わり、全部計算し直しになる デコーダだけの構造が、ここで効いている（問1）

新しいトークン1つぶんの計算の中身

各層で、新しい位置について次を行う

1. 新しい位置のQ・K・Vを計算する（1位置ぶん）
2. 自分のQと、キャッシュにある過去の全位置のKを照合し、対応するVを集める
3. FFNを通す（1位置ぶん）
4. 自分のK・Vをキャッシュに追加する

過去の位置を「参照」はするが、過去の位置を「計算し直す」ことはしない キャッシュが保存しているのは、手順2で必要なKとVである

<img src="https://class.west.sd.keio.ac.jp/dataai/text/llmu-15-kv-reuse.png" width=760>

速くなる理由

入力がT個のとき、1トークンを生成する計算量は次のようになる

|  | キャッシュなし | キャッシュあり |
| :--: | :--: | :--: |
| FFNを通す位置の数 | T個（全位置） | 1個 |
| Attentionの照合 | T×T組 | 1×T組 |

- FFNと各種の射影がT倍から1倍になるので、理屈の上ではTにほぼ比例して速くなる 入力が1400トークンなら、最大で千倍規模の差があり得る
- 実測（6.3倍）がそこまで開かないのは、小さなモデルでは1回の呼び出しの固定費が大きいためである
  - GPUが遅いほど、計算量の差がそのまま時間に出る

## 問7  「モデルを1回呼ぶ」の「1回」とは何か

**答え**

「1回」には少なくとも4つの意味があり、混同すると話がかみ合わなくなる

| 「1回」の意味 | 内部で起こり得ること |
| --- | --- |
| `forward`（1回の前向き計算） | 与えられた位置の表現と出力分布を計算する 入力の処理では多数の位置を一度に扱う |
| `generate`（1回の生成要求） | 停止するまで、前向き計算とトークンの選択を繰り返す |
| 1回のAPI要求 | 多数のトークン生成を含み、サービスによっては推論やツール実行も含む |
| 画面上の1回の回答 | 複数のモデル要求、検索、計算、要約をまとめて表示することもある |

**説明**

API（Application Programming Interface）は、ソフトウェア同士の呼び出し口である APIを1回呼んだことから、モデルの前向き計算も1回だったとは判断できない
- 同じように、画面に1つの回答が出たことから、モデルが1回しか動いていないとも言えない

前の節の自作ループは、`generate` を1回呼ぶあいだに前向き計算が64回起きることを示した

## 問8  KVキャッシュとは何か  計算を速くする以上の働きがあるのか

**答え**

KVキャッシュは、**過去のトークンについて計算したAttention用の値（キー K とバリュー V）を、層ごとに保存しておく仕組み**
- 新しいトークンを処理するとき、過去のぶんを計算し直さずに済む

働きは「再計算を省くこと」であり、それ以上ではない キャッシュを使っても使わなくても、答えは原理上同じである（前の節で確かめた）
- 隠れた知識の倉庫ではない

**説明**

Attentionでは、新しい位置の問い合わせ $Q$ を、過去の位置の $K$ と照合し、対応する $V$ を集める 標準的な因果的Attentionでは、過去の位置の表現は未来のトークンに影響されないので、一度計算した $K$ と $V$ をそのまま再利用できる（Hugging Face「How caching works」）

- **生成したトークン列**は離散的な記録、**KVキャッシュ**は文脈を反映した数値の表現である 形式は違うが、決定的な計算では、キャッシュの中身は「重み」と「処理済みのトークン列」から決まる
- 保存するのは、Attentionに必要な K と V だけである 過去のあらゆる一時変数や、全層の計算過程を丸ごと保存するわけではない
- 標準の生成では、過去のトークンを後から書き換えることはしない 「先ほどの値を修正する」という新しいトークンを追加して、後続を変えることはできる

キャッシュの大きさは、式で見積もれる 1トークンあたり、各層でKとVを持つので、層数$L$、KVヘッド数$n_{\text{kv}}$、ヘッドの次元$d_{\text{head}}$、1要素のバイト数$b$とすると、

$$
\text{bytes}_{\text{token}} = 2\times L\times n_{\text{kv}}\times d_{\text{head}}\times b
$$

となる
- これに、文脈の長さ $T$ を掛けたものが、キャッシュ全体の大きさになる

### 確かめる ― KVキャッシュの大きさを、式と実測で突き合わせる

何を実装しているか

- 上の式で、Qwen3-0.6B（float32、$b=4$）の $T$ トークン分のキャッシュの大きさを計算し、前の節で作ったキャッシュ（実際のテンソル）の大きさと**完全に一致する**ことを `assert` で確かめる
- 同じ式で、大きなモデル（層数と次元を仮定した値）の文脈長ごとのキャッシュを見積もる

実装の肝

- 「KVヘッド数」は注意ヘッド数ではない このモデルは注意ヘッド16に対しKVヘッド8である（Grouped-Query Attention） 注意ヘッド数で計算すると2倍に見積もってしまう

結果の読み方

- キャッシュは文脈の長さに比例して増える 長い文脈を保持するほど、重みとは別にGPUメモリを使う 実際の運用で文脈長や同時利用者数が制約される主な理由がこれである

観測上の限界

- キャッシュの量子化、ページ管理（PagedAttention）、プレフィックスの共有などの工夫は含んでいない 大きなモデルの値は、層数と次元を仮定した試算である

````python
# KVキャッシュの大きさを、式で見積もって実測と照合する
def kv_bytes(n_layers, n_kv_heads, head_dim, n_tokens, bytes_per_elem):
    return 2 * n_layers * n_kv_heads * head_dim * bytes_per_elem * n_tokens


T = ids.shape[1] + N_NEW - 1                       # 直前のループで、キャッシュに取り込んだトークン数
predicted = kv_bytes(cfg.num_hidden_layers, cfg.num_key_value_heads, cfg.head_dim, T, 4)
measured = sum(layer.keys.numel() * layer.keys.element_size() + layer.values.numel() * layer.values.element_size()
               for layer in cache.layers)
assert cache.get_seq_length() == T and predicted == measured
print(f"T = {T} トークン: 式 {predicted:,} バイト / 実測 {measured:,} バイト → 完全に一致")

print("1トークンあたり:", f"{kv_bytes(cfg.num_hidden_layers, cfg.num_key_value_heads, cfg.head_dim, 1, 4) / 1024:.0f} KiB（float32）")

# 試算: 層数・KVヘッド数・ヘッド次元を仮定した、大きなモデルの文脈長ごとのキャッシュ（半精度、b=2）
print("\n試算（半精度、1リクエスト分）:")
for name, (L_, kv_, hd_) in {"7B級（32層、KV8、128）": (32, 8, 128), "70B級（80層、KV8、128）": (80, 8, 128)}.items():
    row = [f"{kv_bytes(L_, kv_, hd_, t, 2) / 2 ** 30:6.2f} GiB（{t // 1000}k）" for t in (8_000, 32_000, 128_000)]
    print(f"  {name}: " + " / ".join(row))
````

結果の読み方

- 式と実測は完全に一致する（`T=68` トークンで 15,597,568 バイト、1トークンあたり224 KiB）
- 試算では、70B級（層数80、KVヘッド8）の半精度で、128kトークンの文脈なら、1リクエストあたり約39 GiBになる
  - 重みとは別の費用であり、同時に処理できる利用者数を制約する
- 文脈の窓（コンテキストウィンドウ）には上限がある 入力、途中結果、回答が長くなると、この制約を受ける
  - 推論モデルでは、非表示の推論トークンも予算と文脈の制約に関係する（問13）

## 問9  途中の計算結果は、どこに置かれているのか

**答え**

一番わかりやすい置き場所は、**生成したトークン列そのもの**である 例えば

```text
24 × 17 = 408
38 × 26 = 988
408 + 988 = 1396
1396 − 75 = 1321
```

と書き出せば、後半の計算は先に書いた `408` や `988` を参照できる

紙に途中式を書くのと同じで、頭の中（モデルの内部）だけで全部を覚えておく必要がなくなる
- これを Scratchpad（作業用紙）や CoT（Chain of Thought：思考の連鎖）と呼ぶ（Nye et al., 2021、Wei et al., 2022）

**説明**

生成されるトークンは、会話の言葉だけではない
- 数字、演算記号、変数名、プログラム、表、ツールへの要求も書ける

今の作業状態を $S_t$ とすると、概念的には

$$
S_t \longrightarrow \text{次の操作・中間結果の生成} \longrightarrow S_{t+1}
$$

という遷移を繰り返せる（$S_t$ が内部に1個の変数として存在する、という意味ではない）

途中を書き出すことには、2つの役割がある

- **作業の分解**: 最初から `1321` を出す課題と、`408`、`988`、`1396` が文脈にある状態で最後の引き算をする課題では、使える情報が違う
- **追加の計算の確保**: 次の判断までに、モデルを何度も使える

ただし、長く説明すれば必ず良くなるわけではない
- 誤った前提や不要な説明を増やすと、正解から遠ざかることもある
- 効果はモデル、学習、課題、与え方に依存する

「同じ重みで、なぜ毎回違う手順を実行できるのか」は、普通のプログラムと同じ考え方で理解できる
- プログラムも、実行中に自分を書き換えなくても、変数の値に応じて動作を変える
- モデルも、入力と途中状態が変われば、同じ重みから違う出力が出る 再利用されるのは**同じ変換規則**であって、同じ結果ではない
- ただし、標準のTransformerに明示的な命令ポインタや型付きの変数があるわけではなく、必要な状態と操作の表現を学習できるかは別の問題である

限界もある
- 1つの誤ったトークン（例えば `406`）が後続の前提になると、誤りがそのまま増幅する
- 途中の値を忘れる、方針を変えても整合がとれない、ということも起きる
- だから、手順の生成で**できることの範囲は広がる**が、正しさを確かめるには検証や外部実行（問18、問25）が要る 問12で、同じ問題を使って測る

## 問10  生成を繰り返すために、外側に管理プログラムが必要か

**答え**

要らない

トークンを1つ足してもう一度モデルを通すだけなら、普通の生成エンジン（`generate`）のループで済む 外側のエージェントが「次の手順を考えよ」と毎回指示し直す必要はない

ただし、**ツールを実行して結果を戻す**ような繰り返しは、生成エンジンの外側のプログラム（ハーネス、問18）が行う
- つまり、繰り返しには2種類ある

- 1回の生成要求の中で行う、次トークン生成の繰り返し
- 外側のプログラムがモデル要求を繰り返す、システムレベルの繰り返し

これを混同すると、「思考が長いなら、別の管理モデルが毎段階で呼び直しているはずだ」という誤解が生じる

**説明**

2つのループを擬似コードで分けて示す そのまま動くプログラムではなく、責任の範囲を示すものである

通常の生成ループ（1回の `generate` の中）

```python
logits, cache = model.prefill(prompt_tokens)      # 入力全体を1回で処理する
generated = []
for step in range(max_new_tokens):
    token = select_next_token(logits)             # 貪欲法、または確率に従う抽選
    if is_end_token(token):
        break
    generated.append(token)
    presentation.receive(token)                   # 表示する・しないは、このループと独立
    logits, cache = model.decode(token, cache)    # 過去のK・Vは cache から再利用
```

このループだけでも、同じTransformerを何度も使っている
- `presentation` が画面に出さなくても、生成列への取り込みを止めなければ、そのトークンは後続の計算に影響する

ツールを使うループ（ハーネス、問18で実装する）

```python
context = make_initial_context(user_request)
for round_index in range(max_rounds):
    event = model.generate_until_event(context, tools)   # この中で上の生成ループが何度も進む
    if event.kind == "final":
        return show(validate(event))
    if event.kind == "tool_call":
        check_schema_and_permission(event)                  # 要求と、実行の許可は別
        result = execute_in_restricted_environment(event)
        context = append(context, event, result)            # 結果はデータとして戻す
```

「同じ重みを何度も使うだけで、なぜ新しいことができるのか」という疑問には、加算器のたとえで答えられる
- 同じ加算器に違う数を入れれば違う結果が出るが、回路は変わらない
- モデルも、生成済みの手順や中間値が次の入力に加われば、同じ重みが違う処理をする
- ただし、その状態を正しく読んで次の操作を正しく選ぶ保証は別の問題である

「それなら、Transformerは再帰型ネットワーク（RNN）なのか」については、**1回の前向き計算の構造**と、**それを繰り返す生成システムの構造**を分けて答える
- モデル本体の中に任意回数のループがあるわけではない
- 一方、生成システム全体には、出力を次の入力に戻す時間方向の繰り返しがある
- 隠れ状態を内部に直接戻す特殊な構造（問34）は、さらに別である

## 問11  温度を下げて同じ答えが出れば、その答えは正しいのか

**答え**

正しいとは限らない

温度（temperature）は、モデルが出す確率分布の「とがり具合」を変えるつまみで、正しさとは関係がない
- 温度を下げれば同じ答えが再現されやすくなるが、誤った候補に一番高い確率がついていれば、同じ誤答が毎回出る

**「同じ答えが再現される」と「その答えが正しい」は別である**

**説明**

スコア $z$ から確率を作るとき、正の温度 $\tau$ で $\operatorname{softmax}(z/\tau)$ とする
- $\tau$ が小さいほど最大の候補に集中し、大きいほど平らになる 貪欲法は $\tau\to0$ の極限で、常に最大確率の候補を選ぶことに相当する

「決定的に選べば、正しさも決定的になるのか」という疑問にも同じ答えになる
- 同じ条件で同じ出力になることと、出力が正しいことは別である 再現性を言うには、計算の実装や数値精度もそろえる必要がある

もちろん、温度を上げれば、候補として低い単語を選び続けるため、あげすぎると回答はどんどん劣化する

### 確かめる ― 温度で分布はどう変わるか

何を実装しているか

- 0.6Bのモデルで、`The capital of France is` の次トークンのスコア $z$ を取り出し、$\tau=0.2,\ 0.7,\ 1.0,\ 2.0$ で確率を計算して、上位3候補とエントロピーを表示する
- $\tau=1$ で2000回抽選し、頻度が理論の確率に近いことを確かめる

実装の肝

- 確率の和が1であること、エントロピーが $\tau$ とともに単調に増えることを `assert` で確かめる

結果の読み方

- $\tau$ が小さいほど、上位の候補の確率が1に近づく この例では最大の候補が正解（Paris）だが、それはこの例の性質であって、一般の保証ではない

観測上の限界

- 1つの文脈の、1ステップの分布を見ただけである

````python
# 温度ごとの分布（上位3候補とエントロピー）、サンプリング頻度との照合
enc = tok_s("The capital of France is", return_tensors="pt").to(device)
with torch.no_grad():
    z = small(**enc).logits[0, -1]

entropies = []
for tau in (0.2, 0.7, 1.0, 2.0):
    p = torch.softmax(z / tau, dim=-1)
    assert abs(p.sum().item() - 1) < 1e-4
    top = torch.topk(p, 3)
    entropy = -(p * torch.log(p.clamp_min(1e-12))).sum().item()
    entropies.append(entropy)
    print(f"τ={tau:<4} " + "  ".join(f"{tok_s.decode(i)!r}:{v:.3f}" for v, i in zip(top.values.tolist(), top.indices.tolist()))
          + f"   エントロピー {entropy:.2f}")
assert all(a < b for a, b in zip(entropies, entropies[1:]))

p1 = torch.softmax(z, dim=-1)
draws = torch.multinomial(p1, 2000, replacement=True)
best = p1.argmax().item()
print(f"\nτ=1: 最大確率の候補 {tok_s.decode(best)!r} の理論の確率 {p1[best]:.3f}、2000回のサンプリングでの頻度 {(draws == best).float().mean():.3f}")
````

結果の読み方

- ` Paris` の確率は、$\tau=0.2$ で1.000、$\tau=0.7$ で0.936、$\tau=1.0$ で0.657、$\tau=2.0$ で0.028と下がり、エントロピーは0.00から9.74に増える
- $\tau=1$ で2000回抽選すると、` Paris` の頻度は0.652で、理論の確率0.657に近い
- 温度を下げれば「毎回同じ答え」になるが、それは「答えが正しい」ことの保証ではない

# 「答えだけ」と思考過程

## 問12  「答えだけ示して」と指示すると、モデルは考えなくなるのか

**答え**

ならない

「答えだけ示して」は、**画面に何を出すか**の指定であって、**どれだけ計算するか**の指定ではない

実際に試すと、「答えだけ」と指示しても、モデルは途中の計算を書いてから答えることが多い
- 本当に途中を書かせないようにすると（応答の書き出しを「答え: 」に固定する）、手順の多い問題は解けなくなる
- そして、同じ途中計算を**表示だけ隠しても**、答えは変わらない
- 途中の計算を書かせたあと、**別の呼び出しで答えだけにさせても**、1回目が答えを出していれば、答えは変わらない 画面は短くなるが、計算は1回目で済んでいる

**説明**

<img src="https://class.west.sd.keio.ac.jp/dataai/text/llmu-06-answer-only.png" width=760>

上の図のように、同じ短い画面表示でも、内部の経路は何通りもある

| 条件 | 何が起きるか | 正答率について言えること |
| --- | --- | --- |
| A 本当に直接答える | 入力から答えのトークンを直接生成する | 手順の多い課題で不利になり得る |
| B 途中を生成する | 計算を書いてから答える | 中間生成の効果が出る |
| C Bの表示だけ隠す | 計算は行い、表示だけ省く | 同じ生成結果なら、答えは変わらない |
| D Bの応答を、別呼出しで答え抽出 | 1回目で計算を書き、2回目はその出力を読んで答えだけを書く | 2回呼び出すので、計算量は増える |
| E 非表示の推論トークン | 内部で推論を生成し、最終回答だけ表示する | 短い表示から計算量は推定できない（問13） |
| F 外部ツールで計算 | 計算プログラムの結果を短く表示する | ツールの能力と、正しい立式に依存する（問18） |

この表を、**同じ問題集で実際に測る** 問題は3種類で、それぞれ4つの水準（難しさ）を持つ

- **整数の加算**: 桁数を 2、4、8、16 と変える
- **条件分岐を含む状態更新**: 「更新前の $x$ が偶数なら $x=(ax+b)\bmod 1009$、奇数なら $x=(ax+c)\bmod 1009$」という操作を 4、16、64、256 回続ける 各ステップが前の結果に依存するので、**待ちの長さを段数で変えられる**
- **資料を使う計算**: 架空の仕様書（4、12、32、64件）から、製品と改訂が一致する資料の係数を選んで計算する（問22で使う）

問題と正解は別に持ち、正解はモデルに見せない 答えの判定は、正解の整数と文字列として完全に一致するかどうかである

### 確かめる ― 問題と正解を別々に作る

何を実装しているか

- 3種類×4水準×5問＝60問を乱数から作る 問題は `questions`、正解は `gold`、架空の資料は `corpora` に別々に入れる

実装の肝

- 状態更新の問題は、作るときに実際に計算して、正解と各ステップの途中値（`trace`）を持つ 途中値は、どのステップで誤ったかを調べるのに使える
- 正解は十進整数の**文字列**で持つ 浮動小数点に変換すると、桁の多い数で値が変わる

結果の読み方

- 例として、加算の1問とその正解を表示する

観測上の限界

- 合成問題である 正解しても、一般的なアルゴリズムの獲得や、任意の長さへの汎化は示せない
- 状態更新の列は待ちを含むが、**もっと短い別の解法が存在しない**ことを証明する課題ではない

````python
# 教材用の合成問題を作る（問題・正解・架空資料を別々に持つ）
def make_benchmark(seed=20261011, per_level=3):
    rng = random.Random(seed)
    questions, gold, corpora = [], {}, {}

    def add(qid, family, level, prompt, answer, trace=None):
        questions.append({"id": qid, "family": family, "level": level, "problem": prompt,
                          "direct_instruction": "最後に求めた整数の答えだけを示してください。",
                          "scratchpad_instruction": "途中の値を計算して確認し、最後の行を『答え: 整数』としてください。"})
        gold[qid] = {"answer": str(answer), "trace": trace}      # 正解は、十進整数の文字列で持つ

    for digits in (2, 4, 8, 16):
        for j in range(per_level):
            a, b = rng.randrange(10 ** (digits - 1), 10 ** digits), rng.randrange(10 ** (digits - 1), 10 ** digits)
            add(f"add-{digits}-{j}", "integer_addition", digits, f"整数 {a} と {b} の和を計算してください。", a + b)

    for steps in (4, 16, 64, 256):
        for j in range(per_level):
            modulus, start = 1009, rng.randrange(1009)
            value, lines, trace = start, [], []
            for i in range(steps):
                a, b, c = rng.randrange(2, 10), rng.randrange(1, 100), rng.randrange(1, 100)
                lines.append(f"{i + 1}: 更新前のxが偶数なら x=({a}*x+{b}) mod {modulus}、"
                             f"奇数なら x=({a}*x+{c}) mod {modulus} とする。")
                value = (a * value + (b if value % 2 == 0 else c)) % modulus
                trace.append(value)
            prompt = (f"状態xの初期値は{start}です。以下を上から順に一度ずつ実行し、"
                      "最後のxを求めてください。modは0以上の剰余です。\n" + "\n".join(lines))
            add(f"state-{steps}-{j}", "conditional_state_updates", steps, prompt, value, trace)

    for size in (4, 12, 32, 64):
        for j in range(per_level):
            docs = [{"document_id": f"spec-{size}-{j}-{i}", "product": f"装置-{i // 2:03d}", "revision": 1 + i % 2,
                     "a": rng.randrange(10, 50), "b": rng.randrange(10, 50)} for i in range(size)]
            sel = rng.choice(docs)
            na, nb, red = rng.randrange(10, 50), rng.randrange(10, 50), rng.randrange(10, 100)
            qid = f"rag-{size}-{j}"
            corpora[qid] = [{"document_id": d["document_id"],
                             "text": (f"架空の仕様書。製品={d['product']}、改訂={d['revision']}。係数A={d['a']}、係数B={d['b']}。"
                                      "この資料は教材用の合成データであり、実在の製品仕様ではありません。")} for d in docs]
            prompt = (f"添付または検索した架空資料から、製品={sel['product']}、改訂={sel['revision']}に一致する係数AとBを選んでください。"
                      f"{na}*A+{nb}*B-{red}を整数として計算してください。"
                      "該当資料が取得できない場合は推測せず、その旨を回答してください。")
            add(qid, "document_grounded_calculation", size, prompt, na * sel["a"] + nb * sel["b"] - red,
                {"source_document_id": sel["document_id"], "A": sel["a"], "B": sel["b"]})
    return questions, gold, corpora

# 先に make_benchmark を呼び出して questions などの変数を定義する
questions, gold, corpora = make_benchmark(seed=20261011, per_level=5)

by_family = defaultdict(list)
for q in questions:
    by_family[q["family"]].append(q)

# ここから下は表示だけ: 作ったデータセットの中身を、3種類から1問ずつ見る（モデルはまだ呼ばない）
FAMILY_NAME = {"integer_addition": "整数の加算（水準＝桁数）",
               "conditional_state_updates": "状態更新（水準＝ステップ数）",
               "document_grounded_calculation": "資料を使う計算（水準＝資料の件数）"}


def show_example(q, max_lines=6):
    g = gold[q["id"]]
    lines = q["problem"].splitlines()
    print(f"■ {FAMILY_NAME[q['family']]}  水準 {q['level']}  問題ID {q['id']}")
    print("  [モデルに渡す] 問題文:")
    for ln in lines[:max_lines]:
        print("      " + ln)
    if len(lines) > max_lines:
        print(f"      …（全部で {len(lines)} 行）")
    print("  [モデルに渡さない] 正解:", g["answer"])
    if q["family"] == "conditional_state_updates":
        print("  [モデルに渡さない] 各ステップ後のxの値:", g["trace"])
    if q["family"] == "document_grounded_calculation":
        docs = corpora[q["id"]]
        print(f"  [条件によって渡す] 架空の資料 {len(docs)} 件  1件目: {docs[0]['text']}")
        print(f"  [モデルに渡さない] 正解の根拠になる資料: {g['trace']['source_document_id']}（係数A={g['trace']['A']}、係数B={g['trace']['B']}）")
    print()


print("作ったデータセット: 3種類 × 4水準 × 5問 ＝", len(questions), "問\n")
for family, items in by_family.items():
    print(f"{FAMILY_NAME[family]}: 水準 {sorted({q['level'] for q in items})} を各5問、計 {len(items)} 問")
print()
for family in by_family:
    show_example(by_family[family][0])
````

````python
import os
os.environ["HF_HUB_DISABLE_XET"] = "1"            # 大きなファイルを通常のHTTPで取得する（Xet方式で止まることがあるため）
# モデルの重みを先にダウンロードする 進みが止まったら自動で取り直し、最後に全ファイルの大きさを照合する
import shutil
import subprocess

from huggingface_hub import HfApi, constants, snapshot_download


def download_with_retry(repo, stall_sec=60, max_tries=20, cache_dir=None, clean=False):
    """repo をダウンロードする 進みが stall_sec 秒止まったら、中断して続きから取り直す
    完了の判定は、配布元のファイル一覧と、手元のファイルの大きさが全部一致することで行う
    clean=True にすると、取得済みのファイルを削除して、最初から取り直す"""
    cache_dir = cache_dir or constants.HF_HUB_CACHE
    model_dir = os.path.join(cache_dir, "models--" + repo.replace("/", "--"))
    if clean and os.path.isdir(model_dir):
        shutil.rmtree(model_dir)
        print("取得済みのファイルを削除した:", model_dir)
    expected = {s.rfilename: s.size for s in HfApi().model_info(repo, files_metadata=True).siblings}   # 配布元のファイル名と大きさ

    def check():
        """(手元の置き場所, 足りないか大きさが違うファイルの一覧) を返す"""
        try:
            path = snapshot_download(repo, cache_dir=cache_dir, local_files_only=True)
        except Exception:
            return None, sorted(expected)
        bad = [f for f, size in expected.items()
               if not os.path.isfile(os.path.join(path, f)) or os.path.getsize(os.path.join(path, f)) != size]
        return path, sorted(bad)

    def progress():
        """進み具合の目安: モデルの置き場所にあるファイルの合計と、ディスクの使用量（どちらかが増えていれば、進んでいる）"""
        files = sum(os.path.getsize(os.path.join(d, f)) for d, _, fs in os.walk(model_dir) for f in fs
                    if not os.path.islink(os.path.join(d, f)))
        return files, shutil.disk_usage(cache_dir if os.path.isdir(cache_dir) else os.path.expanduser("~")).used

    total = sum(expected.values())
    path, bad = check()
    print(f"{repo}: {len(expected)}ファイル、{total / 1e9:.1f} GB  取得が必要なファイル {len(bad)}個")
    # ダウンロードは別のプロセスで行う 止まったときに、そのプロセスだけを終了できる
    # HF_HUB_DISABLE_XET=1 で、大きなファイルを通常のHTTPで取得する（Xet方式は途中で止まることがある）
    code = f"from huggingface_hub import snapshot_download; snapshot_download({repo!r}, cache_dir={cache_dir!r}, max_workers=2)"
    env = dict(os.environ, HF_HUB_DISABLE_XET="1", HF_HUB_DISABLE_PROGRESS_BARS="1")
    t0, attempt = time.time(), 0
    while bad and attempt < max_tries:
        attempt += 1
        proc = subprocess.Popen([sys.executable, "-c", code], env=env, stdout=subprocess.DEVNULL, stderr=subprocess.PIPE, text=True)
        base, last, last_change, last_print = progress(), progress(), time.time(), time.time()
        while proc.poll() is None:
            time.sleep(2)
            now = progress()
            if max(now[0] - last[0], now[1] - last[1]) > 1e6:                 # 1MB以上増えていれば、進んでいる
                last, last_change = now, time.time()
            if time.time() - last_print >= 15:
                grown = max(now[0] - base[0], now[1] - base[1])
                print(f"  {attempt}回目: この回に取得した量 {grown / 1e9:.2f} GB（{time.time() - t0:.0f}秒）")
                last_print = time.time()
            if time.time() - last_change > stall_sec:
                proc.kill()
                print(f"  {attempt}回目: {stall_sec}秒進まなかったので中断し、続きから取り直す")
                break
        err = (proc.communicate()[1] or "").strip().splitlines()
        if proc.returncode not in (0, None) and time.time() - last_change <= stall_sec and err:
            print(f"  {attempt}回目: エラーで終了した（{err[-1][:200]}）")
            time.sleep(5)
        path, bad = check()
        if bad and proc.returncode == 0:
            print(f"  {attempt}回目: 終了したが、{len(bad)}個のファイルがそろっていない: {bad[:5]}")
    if bad:
        raise RuntimeError(f"{repo}: ダウンロードが完了しなかった 足りないファイル {bad}")
    got = sum(os.path.getsize(os.path.join(path, f)) for f in expected)
    print(f"ダウンロード完了: {len(expected)}ファイルの大きさが配布元と一致、合計 {got / 1e9:.1f} GB、取得 {attempt}回、{time.time() - t0:.0f}秒")
    for f in sorted(expected, key=expected.get, reverse=True)[:3]:
        print(f"  {f}: {os.path.getsize(os.path.join(path, f)) / 1e9:.2f} GB")
    return path


INSTRUCT = "Qwen/Qwen3-4B-Instruct-2507"
download_with_retry(INSTRUCT)
````

### 確かめる ― 比較に使う道具

何を実装しているか

- 問題をまとめて解かせる
  - `chat_batch`、答えを読み取る関数、条件ごとに正解数を集計する `evaluate` と `summarize` を用意する
- 別の条件の応答をもう一度モデルに読ませて、答えだけにさせる
  - `evaluate_second_call` と、その集計の `summarize_second_call` も用意する（条件Dで使う）

実装の肝

- `chat_batch` の `prefill` に文字列を渡すと、**応答の書き出しをこちらで固定**できる
  - 「答え: 」を固定すれば、モデルは途中式を書く余地がない これが条件Aである
- 条件Bの指示には、付属の指示文に書式の例（`答え: 123`）を足してある
  - 例がないと、モデルが文字どおり「答え: 整数」と書くことがある 条件はすべて同じ指示を使う
- `evaluate_second_call` は、1回目の答え（`src_pred`）、1回目の生成トークン数（`src_tokens`）、2回目が入力として読んだトークン数（`in_tokens`）も記録する
  - 2回目の画面表示が短くても、その前にどれだけ計算したかを後で数えられる
- 生成の上限トークン数（`scratch_budget`）は、途中で打ち切られないように余裕を持たせる
  - 上限に達すると答えがなく不正解になるので、平均トークン数が上限と同じなら打ち切りを疑う

結果の読み方

- 表の各欄は「正解数/問題数（平均の生成トークン数）」である

観測上の限界

- 各条件5問である 1問の差が20ポイントの差になるので、1〜2問の差は誤差の範囲として読む（「規模と統計的な誤差」を参照）

````python
# 比較実験の道具: バッチ生成（応答の書き出しを固定できる）、答えの読み取り、評価と集計
results = []


def chat_batch(prompts, max_new_tokens, system=None, prefill="", batch_size=8, chunk=512):
    """各プロンプトに貪欲法で応答させる prefill を与えると、応答の書き出しを固定できる
    GPUメモリに収まるように、入力と出力の長さに応じて、1回にまとめて処理する問題数を自動で減らす
    長い入力（1024トークン超）は1問ずつ、chunk トークンずつに区切って読み込む（KVキャッシュを使う 問8）"""
    texts = [tok.apply_chat_template(([{"role": "system", "content": system}] if system else [])
                                     + [{"role": "user", "content": p}], tokenize=False, add_generation_prompt=True) + prefill
             for p in prompts]
    lens = [len(tok(t).input_ids) for t in texts]
    outs, i = [], 0
    while i < len(texts):
        # まとめる問題数 n を決める Attentionの作業領域は n×T×T、KVキャッシュは n×(T＋生成数) に比例する（T は最長の入力）
        n = 1
        while n < batch_size and i + n < len(texts):
            T = max(lens[i:i + n + 1])
            if (n + 1) * T * T > 3.2e6 or (n + 1) * (T + max_new_tokens) > 12000:
                break
            n += 1
        enc = tok(texts[i:i + n], return_tensors="pt", padding=True).to(device)
        T, past = enc["input_ids"].shape[1], None
        with torch.no_grad():
            if n == 1 and T > 2 * chunk:
                # 長い入力を一度に読むと、Attentionの作業領域（T×T）がGPUメモリに収まらない
                # 先頭から chunk トークンずつ読み、結果をKVキャッシュにためる 作業領域は chunk×T で済む
                for s in range(0, T - chunk, chunk):
                    past = llm(input_ids=enc["input_ids"][:, s:s + chunk], past_key_values=past, use_cache=True).past_key_values
            g = llm.generate(**enc, past_key_values=past, max_new_tokens=max_new_tokens,
                             do_sample=False, temperature=None, top_p=None, top_k=None)
        for row in g[:, T:]:
            outs.append((tok.decode(row, skip_special_tokens=True).strip(), int((row != tok.pad_token_id).sum())))
        i += n
        del enc, g, past
        torch.cuda.empty_cache()                  # 次のまとまりの前に、使い終えたGPUメモリを返す
    return outs

# ---- 答えの読み取り ----
def first_int(text):
    m = re.search(r"-?\d[\d,]*", text)
    return m.group(0).replace(",", "") if m else None


def last_int(text):
    m = re.findall(r"-?\d[\d,]*", text)
    return m[-1].replace(",", "") if m else None


def answer_line(text):
    """『答え: 整数』の形の最後の行から、整数を読む（形になっていなければ None）"""
    m = re.findall(r"答え\s*[:：]\s*\**\s*(?:整数)?\s*(-?[\d,]+)", text)
    return m[-1].replace(",", "") if m else None


# ---- 評価（1回の呼び出し）----
def evaluate(cond, items, prompts, max_new_tokens, parse, system=None, prefill="", batch_size=8):
    t0 = time.time()
    outs = chat_batch(prompts, max_new_tokens, system, prefill, batch_size)
    for q, (text, n) in zip(items, outs):
        pred = parse(text)
        results.append({"cond": cond, "id": q["id"], "family": q["family"], "level": q["level"],
                        "pred": pred, "ok": pred == gold[q["id"]]["answer"], "n_tokens": n, "text": text})
    print(f"{cond:<10} {len(items):2d}問 {time.time() - t0:5.0f}秒")


# ---- 評価（2回目の呼び出し）: 別の条件の応答を読ませて、答えだけにさせる ----
EXTRACT = ("次の文章は、ある計算問題への解答である。この解答が最終的に出した答えの整数だけを書け。"
           "計算し直してはいけない。\n\n解答:\n")


def evaluate_second_call(cond, src_cond, batch_size=8):
    """src_cond の応答（途中の計算つき）を2回目の呼び出しに渡し、書き出しを『答え: 』に固定して答えだけを書かせる"""
    t0 = time.time()
    src = [r for r in results if r["cond"] == src_cond]
    prompts = [EXTRACT + r["text"] for r in src]
    outs = chat_batch(prompts, 24, prefill="答え: ", batch_size=batch_size)
    for r, p, (text, n) in zip(src, prompts, outs):
        pred = first_int(text)
        results.append({"cond": cond, "id": r["id"], "family": r["family"], "level": r["level"],
                        "pred": pred, "ok": pred == gold[r["id"]]["answer"], "n_tokens": n, "text": text,
                        "src_pred": r["pred"], "src_tokens": r["n_tokens"],        # 1回目の答えと、1回目の生成トークン数
                        "in_tokens": len(tok(p)["input_ids"])})                    # 2回目が入力として読んだトークン数
    print(f"{cond:<10} {len(src):2d}問 {time.time() - t0:5.0f}秒")


# ---- 集計 ----
def summarize(conds, family):
    """条件×水準の表を、「正解数/問題数（平均トークン数）」で表示する"""
    levels = sorted({r["level"] for r in results if r["family"] == family})
    print(f"{family}" + " " * 4 + "".join(f"{c:>18}" for c in conds))
    for lv in levels:
        row = []
        for c in conds:
            rs = [r for r in results if r["cond"] == c and r["family"] == family and r["level"] == lv]
            row.append(f"{sum(r['ok'] for r in rs)}/{len(rs)}（{np.mean([r['n_tokens'] for r in rs]):.0f}）" if rs else "-")
        print(f"  水準 {lv:>3}" + "".join(f"{x:>18}" for x in row))


def summarize_second_call(cond):
    """2回目の呼び出しの条件について、1回目と答えが一致した件数と、平均トークン数を表示する"""
    rs_all = [r for r in results if r["cond"] == cond]
    print(f"1回目: {sum(r['src_pred'] == gold[r['id']]['answer'] for r in rs_all)}/{len(rs_all)}問正解 → "
          f"{cond}: {sum(r['ok'] for r in rs_all)}/{len(rs_all)}問正解"
          f"（1回目と同じ答え: {sum(r['pred'] == r['src_pred'] for r in rs_all)}/{len(rs_all)}）")
    for family in dict.fromkeys(r["family"] for r in rs_all):
        for lv in sorted({r["level"] for r in rs_all if r["family"] == family}):
            rs = [r for r in rs_all if r["family"] == family and r["level"] == lv]
            print(f"  {family:<26} 水準 {lv:>3}  正解 {sum(r['ok'] for r in rs)}/{len(rs)}"
                  f"  1回目の生成 {np.mean([r['src_tokens'] for r in rs]):.0f}"
                  f"  2回目の入力 {np.mean([r['in_tokens'] for r in rs]):.0f}"
                  f"  2回目の生成 {np.mean([r['n_tokens'] for r in rs]):.0f}")


# ---- 実験の設定 ----
FORMAT_HINT = "（書式の例: 最後の行は『答え: 123』のように、数字だけを書く）"
STATE_LEVELS = (4, 16, 64)                                  # 256 は、出力が非常に長くなるので省く
RUN_STATE = (4,)                                            # A0・B・Fで実行する状態更新の水準 16と64は、生成が長くなるので、次の節で結果だけ示す
sets = {"integer_addition": by_family["integer_addition"],
        "conditional_state_updates": [q for q in by_family["conditional_state_updates"] if q["level"] in STATE_LEVELS]}


def scratch_prompt(q):                                       # 条件Bの指示（付属の指示文に、書式の例を足したもの）
    return q["problem"] + "\n" + q["scratchpad_instruction"] + FORMAT_HINT


def scratch_budget(q):                                       # 生成の上限トークン数（途中で打ち切られないように、余裕を持たせる）
    if q["family"] == "integer_addition":
        return 1200
    return 200 + 200 * q["level"]                            # 状態更新は、1ステップに100〜170トークン程度を使うので、余裕を持たせる
````

### 確かめる ― 条件A（直接）、A0（指示だけ）、B（途中を書く）

何を実装しているか

- 条件A: 書き出しを `答え: ` に固定し、数トークンだけ生成させる
  - 途中式を書く余地がない
- 条件A0: 「答えだけを示してください」と**指示するだけ**で、書き出しは固定しない
- 条件B: 「途中の値を計算して確認し、最後の行を『答え: 整数』としてください」と指示する

実装の肝

- 「答えだけ」と**指示する**こと（A0）と、「答えだけを生成させる」こと
  -（A）は別である 両方を測ることで、指示が表示の指定であって計算の指定ではないことを確かめる
- 条件Aは出力の最初の整数、A0は最後の整数、Bは「答え: 」の行の整数を答えとして読む
- 状態更新の16ステップと64ステップは生成が長くなるので、ここでは4ステップまでを実行し、残りは後の節で結果だけ示す

結果の読み方

- 各行は、条件、問題数、所要時間である

観測上の限界

- 貪欲法で1回ずつ実行している 抽選を使うと結果は変わる

````python
# 条件A（直接回答: 書き出しを固定）、条件A0（指示だけ）、条件B（中間生成）を実行する
for family, items in sets.items():
    evaluate("A直接", items, [q["problem"] + "\n" + q["direct_instruction"] for q in items], 24, first_int, prefill="答え: ")
    for lv in sorted({q["level"] for q in items}):
        it = [q for q in items if q["level"] == lv]
        if family == "integer_addition" or lv in RUN_STATE:
            evaluate("A0指示だけ", it, [q["problem"] + "\n" + q["direct_instruction"] for q in it], scratch_budget(it[0]), last_int)
            evaluate("B中間生成", it, [scratch_prompt(q) for q in it], scratch_budget(it[0]), answer_line)
````

### 確かめる ― 条件C（表示だけ隠す）

何を実装しているか

- 条件Bの生成結果から**最後の行だけ**を取り出して「画面に見える部分」にする
  - モデルは呼ばない

実装の肝

- 再生成しないことが要点である
  - 問題をもう一度解かせると、表示以外も変わってしまう（Bの応答をモデルに読ませて答えだけにさせる方法は、次の条件Dで試す）
- 画面に見える最後の行から読んだ答えが、条件Bの答えと一致することを全件で `assert` する

結果の読み方

- BとCの正解数は必ず同じになる 例として、生成したトークン数と、画面に見える1行を並べる

観測上の限界

- これは表示の層（プログラム）で途中を隠す場合である
  - 「答えだけと指示して生成し直す」（A0）とも、「モデルをもう一度呼んで答えだけにさせる」（D）とも別の実験である

````python
# 条件C: Bと同じ生成結果の、最後の行だけを表示する（再生成しない）
def display_last_line(text):
    return [ln for ln in text.splitlines() if ln.strip()][-1].strip()

b_rows = [r for r in results if r["cond"] == "B中間生成"]
for r in b_rows:
    shown = display_last_line(r["text"])
    # 画面に見える最後の行から読んだ答えが、Bの答えと一致する（表示を変えても、計算は変わらない）
    assert answer_line(shown) == r["pred"]
    results.append({**r, "cond": "C表示だけ隠す", "text": shown, "n_tokens": r["n_tokens"]})
n_ok = sum(r["ok"] for r in b_rows)
print(f"B: {n_ok}/{len(b_rows)}問正解 → C（最後の行だけ表示）: {n_ok}/{len(b_rows)}問正解（全件で答えが一致）")
print()
ex = next(r for r in b_rows if r["family"] == "integer_addition" and r["level"] == 4)
print(f"例 {ex['id']}: 生成 {ex['n_tokens']} トークン → 画面に見えるのは「{display_last_line(ex['text'])}」だけ")
````

### 確かめる ― 条件D（もう一度呼び出して、答えだけにする）

何を実装しているか

- 条件Bの応答（途中の計算つき）を、**そのまま2回目の呼び出しの入力に入れ**、「答えの整数だけを書け」と指示する
  - 利用者に見せるのは、2回目の出力だけである
- これは、手順つきで一度答えさせてから、別のプロンプトで答えだけにまとめさせる、という実際によく使われる方法である
  - Zero-shot CoT（Kojima et al., 2022）も、1回目で推論を書かせ、2回目で答えだけを取り出す

実装の肝

- 2回目の指示には「計算し直してはいけない」と書く 2回目の役割は**書き写し**であって、計算ではない
- 2回目の書き出しは、条件Aと同じ「答え: 」に固定する 生成も数トークンである **条件Aとの違いは、入力に途中の計算があるかどうかだけ**である
- 2回目の答えが条件Bの答えと同じかどうかを、全件で数える

結果の読み方

- 2行目は、1回目（条件B）とDの正解数と、答えが一致した件数である
- 続く行は、水準ごとの正解数と、平均のトークン数（1回目の生成、2回目の入力、2回目の生成）である
- 最後の行は、1問の例である 1回目に生成したトークン数と、2回目の画面表示を並べる

観測上の限界

- 条件C（プログラムで最後の行を切り出す）と違い、2回目もモデルが書くので、**書き写しを誤る可能性がある**
  - 一致は保証されないので、`assert` ではなく件数を数えている
- 2回目は、1回目の計算の誤りを直さない Bが誤った答えを出した問題は、Dも同じ誤答になる
- 1回目が上限で打ち切られて「答え: 」の行が無いときは、BとDの答えが分かれる Bは答えなしになるが、Dは途中に書かれた数字を答えとして読み取る

````python
# 条件D: Bの応答（途中の計算つき）を、2回目の呼び出しに渡して、答えだけにさせる
evaluate_second_call("D別の呼び出し", "B中間生成")
summarize_second_call("D別の呼び出し")

ex = next(r for r in results if r["cond"] == "D別の呼び出し" and r["family"] == "conditional_state_updates")
print(f"\n例 {ex['id']}: 1回目 {ex['src_tokens']} トークン生成 → 2回目の画面表示「答え: {ex['text']}」")
````

### 確かめる ― 条件F（外部のプログラムで計算する）

何を実装しているか

- モデルに、問題を**Pythonのプログラムに直させる** 実行は別のプロセスに任せ、標準出力の最後の整数を答えとして読む
  - これは PAL（Program-aided Language Models）の考え方である（Gao et al., 2022）

実装の肝

- 実行は `subprocess` で起動した別のPythonプロセスで行い、10秒で打ち切る
- コードブロックが無い、`print()` が無い、実行時に例外が出る、のどれでも出力は空になり、不正解として数える 指示には書式の例（`print(3 + 4)`）を入れてある
- 状態更新では、モデルは各ステップの係数を**書き写す**必要がある 演算はPythonが行うので、誤るとすれば書き写しか立式である

結果の読み方

- 最後に6つの条件を並べた表を表示する
  - 外部計算の生成トークン数は、**プログラムの長さ**である

観測上の限界

- `run_python` は安全な隔離ではない ファイルやネットワークへのアクセスを制限していないので、信頼できないモデルや入力に対してこの方法でコードを実行してはいけない
  - 実運用では、制限された実行環境を使う

````python
# 条件F: 問題をPythonのプログラムに直させ、別のプロセスで実行して答えを得る
PAL_SYSTEM = ("問題をPythonのプログラムに直せ。出力は```pythonで始まるコードブロック1つだけとし、説明とコメントは書かない。"
              "答えは必ず print() で表示し、整数だけを出力せよ。例: ```python\nprint(3 + 4)\n```")


def extract_code(text):
    m = re.search(r"```(?:python)?\n(.*?)```", text, re.S)
    return m.group(1) if m else None


def run_python(code, timeout=10):
    """別のPythonプロセスで、10秒を上限に実行する（授業用の簡易な実行環境であり、安全な隔離ではない）"""
    try:
        r = subprocess.run([sys.executable, "-I", "-c", code], capture_output=True, text=True, timeout=timeout)
        return r.stdout.strip()
    except subprocess.TimeoutExpired:
        return ""


def parse_pal(text):
    code = extract_code(text)
    out = run_python(code) if code else ""
    return last_int(out) if out else None


for family, items in sets.items():
    items = [q for q in items if q["family"] == "integer_addition" or q["level"] in RUN_STATE]
    evaluate("F外部計算", items, [q["problem"] for q in items], 2500 if family != "integer_addition" else 600, parse_pal, system=PAL_SYSTEM)

for family in sets:
    print()
    summarize(["A直接", "A0指示だけ", "B中間生成", "C表示だけ隠す", "D別の呼び出し", "F外部計算"], family)
````

結果の読み方

各条件5問、貪欲法で、次のようになる 表の各欄は「正解数/問題数（平均の生成トークン数）」である

- **A 直接回答**: 加算は2桁と4桁で5/5、8桁と16桁で3/5である
  - 生成は3〜18トークンで、答えの桁数とほぼ同じである
  - 状態更新は**4ステップでも0/5**である 途中を書く余地がないと、手順の多い問題は解けない
- **A0 指示だけ**: 「答えだけ」と指示しただけでは、モデルは**途中の計算を書いてから答える** 状態更新の4ステップでは平均640トークンの手順を書いて5/5である
  - 加算でも、答えの桁数より多いトークン（17〜72）を生成している 式を書いてから答えるためである
  - 指示は表示の指定であって、計算を禁じる指定ではない この条件では計算は画面に見えていて、隠れてはいない
- **B 途中を書く**: 加算は5/5、5/5、3/5、4/5、状態更新の4ステップは5/5（763トークン）である
  - 8桁と16桁の加算では、BはA0より10倍以上のトークンを使ったのに、正解数は同じか少ない（8桁はA0が5/5、Bが3/5）
  - 8桁の加算の平均は907トークンで、上限の1200トークンに近い 途中を長く書きすぎて上限で打ち切られ、「答え: 」の行に届かなかった問題がある 打ち切られた問題は答えなしで、不正解になる
  - **途中を長く書けば常に有利とは限らない**
- **C 表示だけ隠す**: Bと全件で答えが一致する（`assert` が通る） 例えば `add-4-0` は82トークンを生成し、画面には「答え: 7896」の1行だけが見える
  - 表示を隠しても計算の量は変わらない
- **D 別の呼び出しで答えだけにする**: 25問のうち23問で、Bと同じ答えになる 正解数は、Bが22問、Dが23問である
  - 答えが分かれた2問は、どちらも8桁の加算で、Bが上限で打ち切られて「答え: 」の行が無かった問題である Bは答えなしになり、Dは途中に書かれた数字を答えとして読み取った そのうち1問は正解で、Dの正解数が1つ増えた
  - Bが答えを出した23問では、DはすべてBと同じ答えである Bが誤った答えを出した問題は、Dも同じ誤答である **2回目は書き写すだけで、計算の誤りを直さない**
  - 表のDの欄のトークン数は**2回目の生成だけ**（3〜18トークン）で、画面に見えるのもこれだけである
  - その前に1回目の生成（状態更新の4ステップで平均763トークン）があり、2回目はそれを入力として読み直している（平均798トークン）
  - **画面が短くなっても、計算は減らず、むしろ増える**
- **AとDの対比**: どちらも書き出しは「答え: 」で、生成は数トークンである
  - 状態更新の4ステップは、Aが0/5、Dが5/5である 違いは、**入力に途中の計算があるかどうか**だけである
  - 答えを数トークンで書けるかどうかは、それまでに計算が済んでいるかどうかで決まる
- **F 外部計算**: 全水準で5/5である
  - 生成は加算で14〜42トークン、状態更新の4ステップで386トークンである
  - モデルが書いたのはプログラムで、**演算を実行したのはPython**である

解けないときに必ず無回答になるわけではない
- 条件Aの状態更新では、全問でもっともらしい整数が出力された **計算できなくても、もっともらしい誤答を出せる**
- 評価すべきなのは「文字列が出たか」ではなく、正答率や検証の結果である

条件Fの失敗の出方も記録しておく
- 指示に書式の例（`print(3 + 4)`）を入れない場合、モデルが `print()` を書かずに式だけを出力することがあり、2桁と4桁の加算でも不正解になる
- これは演算の誤りではなく**出力形式の誤り**である
- 外部計算の失敗は、演算ではなく、立式、出力形式、書き写しとして現れる

観測上の限界

- 各水準5問である 1問の差が20ポイントの差になるので、「A0がAより良い」「DがBより1問多い」のような小さな差は優劣として読まない

## 問13  推論モデルに「答えだけ」と指示したら、何が起きるのか

**答え**

推論モデル（o1、Gemini thinking、Qwen3-Thinking など）は、最終回答の前に**非表示の推論トークン**を生成するように訓練されている
  - 「答えだけ示して」と指示すると、画面には短い答えだけが出るが、その手前で数千トークンの推論が生成されていることがある

```text
利用者: 答えだけ示して
内部  : 非表示の推論を生成 → 最終回答を生成
画面  : 1321
```

この構成に矛盾はない
  - 画面が4文字でも、内部では長い計算ができる **画面の短さから計算量は推定できない**

**説明**

OpenAIやGoogleの公開仕様では、最終回答の前に内部の推論トークンを生成すること、その生のトークン列を通常は公開しないこと、推論の量を設定で制御できることが説明されている（OpenAI「Reasoning models」、Google「Gemini thinking」） 推論の量は、文脈や料金の制約にも関係する

公開モデルでは `Qwen/Qwen3-4B-Thinking-2507` が、この構成を手元で観察できる例である
  - 推論部分は `</think>` の手前に出力されるので、推論部分と最終回答を分けて数えられる

逆に「推論をしない」設定は何を止めるのか
  - 初期のQwen3には、チャットテンプレートの `enable_thinking=False` で思考用テキストの生成を切り替える仕組みがある
  - これは、思考用トークンを**生成しない**ことであって、各層の内部計算がなくなることではない
  - 直接回答を作るためのAttentionやFFNは変わらず実行される 設定名や動作はモデルと版ごとに違うので、1つの例を全モデルに一般化しない

答えが複数トークンなら、答えを書いているあいだにも生成は繰り返される
- ただし、**答えの最初のトークンを出す前に、まだ出していない後続トークンを使って計算することはできない**
- 「答えが長いから、最初の桁を出す前に十分考えていたはずだ」とは言えない

最も厳しく制限した基本形を考えると、話がはっきりする
- 固定深さの通常Transformer、固定された入力、内部の反復なし、非表示トークンなし、ツールなし、複数候補なし、答えは本当に1トークン
- この場合、答えの分布は入力の1回の前向き計算で決まる
- 「画面に見えないから、任意回数考えたはずだ」と補うことはできない
- 条件Aは、この基本形に近い設定である 逆に、これらの条件を確認できない商用の画面だけを見て、「1回の計算で正答した」とも断定できない（問17）

### 確かめる ― 出力を、思考・回答・ツール呼び出しに分ける

何を実装しているか

- モデルの生の出力（1本の文字列）を、`thinking`、`content`、`tool_calls` の3つに分ける関数 `split_message` を作る

実装の肝

- `</think>` より前を思考、後ろを回答とする `<tool_call>` と `</tool_call>` に挟まれたJSONを、関数名と引数として読む
- 2つの `assert` で、思考と回答の分離、ツール呼び出しの取り出しを確かめる

結果の読み方

- 「分離の確認: OK」と出れば、2つの検査を通っている

観測上の限界

- Qwen3系の書式を前提にしている 別のモデルは別の特殊トークンを使う 実行基盤は、モデルごとにこの変換を用意している（問32）

````python
# 生の出力を、thinking / content / tool_calls に分ける（Qwen3系の書式を前提にした例）
THINKING = "Qwen/Qwen3-4B-Thinking-2507"


def split_message(raw):
    thinking = ""
    if "</think>" in raw:
        thinking, raw = raw.split("</think>", 1)
        thinking = thinking.replace("<think>", "").strip()
    calls = []
    for body in re.findall(r"<tool_call>\s*(\{.*?\})\s*</tool_call>", raw, re.S):
        try:
            d = json.loads(body)
            calls.append({"function": {"name": d["name"], "arguments": d.get("arguments", {})}})
        except (json.JSONDecodeError, KeyError):
            pass
    content = re.sub(r"<tool_call>.*?</tool_call>", "", raw, flags=re.S).strip()
    return {"role": "assistant", "content": content, "thinking": thinking, "tool_calls": calls}


m = split_message('<think>\nまず数値を確認する\n</think>\n\n計算結果は123である')
assert m["thinking"] == "まず数値を確認する" and m["content"] == "計算結果は123である" and m["tool_calls"] == []
m = split_message('<tool_call>\n{"name": "integer_expression", "arguments": {"expression": "1+2"}}\n</tool_call>')
assert m["tool_calls"][0]["function"]["name"] == "integer_expression" and m["content"] == ""
print("分離の確認: OK")
````

## 問14  画面に表示される「思考過程」は、モデルの中身がそのまま見えているのか

**答え**

違う

モデルの中身は、多数の実数が並んだ配列（活性値）である
- 画面に出る日本語の「思考」は、その配列を表示したものではなく、**モデルが文字として生成した推論テキスト**か、その**要約**である

問3のロジットレンズで見たように、途中の層を無理に読んでも、人間に読める文章にはならない
- 「中身は一切見えない」でも「画面の思考が中身そのもの」でもない
- **観測できる数値と、その意味の解釈は別**である

**説明**

<img src="https://class.west.sd.keio.ac.jp/dataai/text/llmu-07-thought-display.png" width=760>

画面に見えるものは、少なくとも5種類に分けられる

| 画面に見えるもの | 正体 | そこから言えないこと |
| --- | --- | --- |
| 「まず条件を整理する…」という文章 | モデルが生成した推論テキスト | 内部計算の全過程が書かれているとは限らない |
| 短く整理された「思考」 | 推論の表示用の要約 | 元の推論と長さ・順番・内容が一対一とは限らない |
| 「資料を確認しています」 | モデルの進捗文やシステムの状態表示 | 検索が完了したことの証明にはならない |
| 検索先、実行コード、戻り値 | ツールや実行基盤の記録 | ツールへの問いが正しかったとは限らない |
| 回答後の「解き方」 | 読者向けに生成した説明 | 回答のときにその順で処理したとは限らない |

推論の要約が生の推論と区別されることは、OpenAIとGoogleの仕様で確認できる

生の推論テキストにも、**忠実性の限界**がある 公開研究では、モデルの答えに影響したヒントが、生成された推論の説明に十分に現れない場合が示されている（Chen et al., 2025）
- 推論テキストは問題解決や監査の手がかりになっても、完全な因果の記録とは限らない
- 「役に立つ情報を含む」ことと「完全で忠実である」ことは別である

研究用の環境なら、活性値やAttentionの値を記録したり、一部の状態を置き換えて出力への影響を調べたりできる
- しかし、数値を取得できることと、それを人間が読める思考として解読できることは別である

関連する疑問に、まとめて答える

- **推論トークンは人間が読める文章か**:
  - 必ずしもそうではない 自然言語、数式、プログラム、記号を含み得る
    - 大事なのは、生成された表現が後続の計算にどう使われるかである
- **途中の手順がもっともらしければ、それが答えの理由か**:
  - そうとは限らない 因果的な寄与を調べるには、入力や内部状態への介入、手順を変えた比較、実行記録が必要である

## 問15  思考が表示されるなら、裏で別のモデルが考えているのか

**答え**

その必要はない

**1つのモデル**が推論用のトークンと最終回答用のトークンを続けて生成し、出力側が `</think>` のような区切りで分けて表示する、という構成で実現できる 問13の実験はまさにこの構成で、動いているモデルは1つである

もちろん、要約やモデルの選択を別の部品で実装する構成もあり得る しかし、**表示されているという事実だけでは、同じモデルか別のモデルか、外側のプログラムが何回介入したかはわからない** 「考えています」と画面に出たことは、別モデルの証拠にならない

**説明**

推論トークンを多く使うモデルも、同じモデルの自己回帰生成だけで多数のトークンを生成できる 別のモデルで候補を作って検証する構成（問35）もあるが、その存在は実装や公開資料で確認すべきである

## 問16  「もう少し考えて」と書けば、計算時間は増えるのか

**答え**

必ずではない

増えるには、その指示を受けて実際に「生成が長くなる」「推論モードが選ばれる」「追加の探索が実行される」という動作が起きる必要がある 問12の条件A0のように、指示がなくてもモデルが途中を書くこともあれば、指示しても書かないこともある

逆に、文章が長くなっても、有効な検証が増えなければ正答率は上がらない 推論の予算を設定できる製品では、自然言語の指示とは別に、設定値を確かめる

**答え**

通常のサービスの画面だけでは、そう結論できない

# 外部の仕組み（ツール・検索・記憶）

## 問18  LLMが検索や計算をできるのは、Transformerがそれをしているのか

**答え**

違う

Transformerの前向き計算には、ネットワークに要求を送る機能も、プログラムを実行する機能もない

LLMがしているのは、「この検索をしてほしい」「この式を計算してほしい」という**要求を文字で書くこと**である

- 要求を受け取った外側のプログラム（ハーネス）が検索や計算を実行し、結果を文字にしてモデルの入力に戻す モデルは、その結果を読んで続きを書く

**説明**

<img src="https://class.west.sd.keio.ac.jp/dataai/text/llmu-08-tool-loop.png" width=760>

上の図が、ツール呼び出し（Function Calling）の流れである

> モデルに使えるツールを知らせる → モデルが呼び出しの要求を返す → ハーネスが検証して実行する → 結果をモデルに返す → モデルが回答か次の要求を生成する

例えば、モデルは次のような要求を生成する

```json
{"tool": "integer_expression", "arguments": {"expression": "24 * 17 + 38 * 26 - 75"}}
```

これはまだ計算結果ではない 要求を受け取ったハーネスが、許可された計算機能に渡して `1321` を得る モデルはその戻り値を見て、説明や回答を作る

**ハーネス（harness）**は、モデルを囲んで実行を支える仕組みの総称で、「モデルの呼び出し、ツールの実行、状態の管理、検証、終了の判断を行う外側のプログラム」を指す

生成エンジン（`generate`）がトークン生成を担当し、ハーネスがそれを使って「検索する」「結果を戻す」「止める」をつなぐ あらかじめ決めた順番で実行するものをワークフロー、モデルが次の処理を選ぶものをエージェントと呼び分ける（Anthropic, 2024）

複雑な計算では、**プログラムを書くことと実行することを分業できる**
- モデルが100万回の繰り返しを1つずつ文字で書く必要はない
- ループを含む短いプログラムを書けば、100万回の繰り返しはPythonが実行する
  - これは「手順が消えた」のではなく、**手順を実行する主体が、トークン生成からプログラムへ移った**ということである
- プログラムを表示しただけでは実行したことにならない 実行の要求、実行環境、終了状態、戻り値を確かめて初めて、実際の計算につながったと言える

Web検索も同じ構図である 通信は検索ツールやブラウザが行い、検索結果がモデルの入力になる 検索エンジンの内部でTransformerが使われることはあっても、索引の管理、外部データへのアクセス、通信の仕組みは別に必要である

### 確かめる ― 最小のハーネスを書き、要求と実行を分ける

何を実装しているか

- **バックエンド**（`HFBackend`）: Hugging Faceのモデルを、Ollamaの `/api/chat` に近い応答の形（`message.content`、`message.thinking`、`message.tool_calls`）で包む
- **ツール**: 整数の式を厳密に評価する `integer_expression`
- **ハーネス**（`run_harness`）: モデルの要求 → ツール名の検査 → 実行 → 結果を `role: "tool"` で履歴に追加、を `tool_calls` がなくなるまで繰り返す

実装の肝

- `integer_expression` は `eval` を使わない 式を構文木に直し、整数の `+ - * // %` だけを許す `__import__('os')` のような式や、巨大な指数は実行しない
  - `assert` で拒否を確かめる
- ハーネスは、未登録のツール名や引数の誤りで例外が出ても止まらず、`エラー: ...` をツールの結果としてモデルに返す
- ループには反復の上限（6回）がある
  - 終わらないモデルがあっても、ハーネスが止める

結果の読み方

- 実行記録（`records`）に、ツール名、引数、結果、成否が残る
  - モデルが何を要求し、ハーネスが何を実行したかを後から追える

観測上の限界

- 授業用の最小の実装である 認証、承認、サンドボックス、時間制限は省いている

````python
# 最小のハーネス: バックエンド（Ollama風の応答形式）、安全な整数計算、ツール実行のループ
import operator


class HFBackend:
    """Hugging Face のモデルを、message.content / message.thinking / message.tool_calls を持つ応答で包む"""

    def __init__(self, model, tokenizer, name, max_new_tokens=400, think=None):
        self.model, self.tok, self.name, self.max_new_tokens, self.think = model, tokenizer, name, max_new_tokens, think

    def chat(self, messages, tools=None, think=None):
        think = self.think if think is None else think                # 既定の思考設定は、バックエンドごとに持つ
        kw = {} if think is None else {"enable_thinking": think}
        text = self.tok.apply_chat_template(messages, tools=tools, tokenize=False, add_generation_prompt=True, **kw)
        enc = self.tok(text, return_tensors="pt").to(device)
        with torch.no_grad():
            out = self.model.generate(**enc, max_new_tokens=self.max_new_tokens, do_sample=False,
                                      temperature=None, top_p=None, top_k=None)
        new = out[0, enc["input_ids"].shape[1]:]
        return {"model": self.name, "message": split_message(self.tok.decode(new, skip_special_tokens=True).strip()),
                "eval_count": int(new.shape[0])}


_OPS = {ast.Add: operator.add, ast.Sub: operator.sub, ast.Mult: operator.mul,
        ast.FloorDiv: operator.floordiv, ast.Mod: operator.mod, ast.USub: operator.neg}


def integer_expression(expression):
    """整数の + - * // % だけを、構文木をたどって厳密に評価する（eval は使わない）"""
    def ev(node):
        if isinstance(node, ast.Constant) and type(node.value) is int:
            return node.value
        if isinstance(node, ast.BinOp) and type(node.op) in _OPS:
            return _OPS[type(node.op)](ev(node.left), ev(node.right))
        if isinstance(node, ast.UnaryOp) and type(node.op) in _OPS:
            return _OPS[type(node.op)](ev(node.operand))
        raise ValueError("整数の + - * // % だけが使える")
    return ev(ast.parse(expression.strip(), mode="eval").body)


def run_harness(backend, request, tools, impls, max_rounds=6, system=None):
    """モデル要求 → 検証 → 実行 → 結果の差し戻し、を終了まで繰り返す"""
    messages = ([{"role": "system", "content": system}] if system else []) + [{"role": "user", "content": request}]
    records = []
    for round_index in range(max_rounds):
        msg = backend.chat(messages, tools=tools)["message"]
        entry = {"role": "assistant", "content": msg["content"]}
        if msg["tool_calls"]:
            entry["tool_calls"] = msg["tool_calls"]
        messages.append(entry)
        if not msg["tool_calls"]:
            return msg["content"], records, messages
        for c in msg["tool_calls"]:
            name, args = c["function"]["name"], c["function"]["arguments"]
            try:
                if name not in impls:
                    raise KeyError(f"未登録のツール: {name}")
                result, ok = str(impls[name](**args)), True
            except Exception as exc:                              # 失敗も、結果としてモデルへ返す
                result, ok = f"エラー: {exc}", False
            records.append({"round": round_index, "tool": name, "arguments": args, "result": result, "ok": ok})
            messages.append({"role": "tool", "tool_name": name, "content": result})
    raise RuntimeError("許可した反復回数内に終了しなかった")


# 実装の検査: 正しい式は厳密に計算し、危険な式・範囲外の式は拒否する
assert integer_expression("24 * 17 + 38 * 26 - 75") == 1321
assert integer_expression("-7 // 2 + 10 % 4") == -2
for bad in ["__import__('os').system('echo x')", "2 ** 100", "1.5 + 2", "(1).real"]:
    try:
        integer_expression(bad)
        raise AssertionError(f"拒否されなかった: {bad}")
    except ValueError:
        pass
print("integer_expression: 正しい式は厳密に計算し、危険な式・範囲外の式は拒否した")
````

実装の肝

  - 最後に最終回答を表示する

観測上の限界

````python
# 題材: 架空の仕様書を引き、係数を使って整数を計算する（ツールの宣言と実体）
SPECS = {
    "装置-007": "架空の仕様書。製品=装置-007、改訂=2。係数A=17、係数B=26。この資料は教材用の合成データであり、実在の製品仕様ではありません。",
    "装置-008": "架空の仕様書。製品=装置-008、改訂=1。係数A=31、係数B=12。この資料は教材用の合成データであり、実在の製品仕様ではありません。",
}


def lookup_spec(product):
    if product not in SPECS:
        raise KeyError(f"仕様書が見つからない: {product}")
    return SPECS[product]


SPEC_TOOLS = [
    {"type": "function", "function": {
        "name": "lookup_spec", "description": "製品名から、架空の仕様書の全文を返す",
        "parameters": {"type": "object", "properties": {"product": {"type": "string", "description": "製品名（例: 装置-001）"}},
                       "required": ["product"]}}},
    {"type": "function", "function": {
        "name": "integer_expression", "description": "整数の式（+ - * // %）を厳密に計算して、整数を返す",
        "parameters": {"type": "object", "properties": {"expression": {"type": "string", "description": "例: 3 * 4 + 5"}},
                       "required": ["expression"]}}},
]
SPEC_IMPLS = {"lookup_spec": lookup_spec, "integer_expression": integer_expression}
SPEC_SYSTEM = "必要なら、ツールを使うこと。計算は、自分で行わず、必ず integer_expression ツールで行うこと。"
SPEC_REQUEST = ("製品=装置-007の仕様書に書かれた係数を使い、Aを24回、Bを38回実行した合計から、補正値75を引いてください。"
                "答えの整数だけを示してください。")

backend = HFBackend(llm, tok, "Qwen3-4B-Instruct-2507")
final, records, messages = run_harness(backend, SPEC_REQUEST, SPEC_TOOLS, SPEC_IMPLS, system=SPEC_SYSTEM)
for r in records:
    print(f"[{r['round']}] {r['tool']}({r['arguments']}) → {r['result'][:60]}{'' if r['ok'] else '（失敗）'}")
print("最終回答:", final)
````

- 最終回答だけを見ても、この経路はわからない

## 問19  ツールを使う仕組みには何が必要か  MCPとは何か

**答え**

「モデルが要求したこと」と「実行してよいこと」は違う だからハーネスには、要求をそのまま実行しないための制御が要る

| 制御 | 必要な理由 |
| --- | --- |
| ツール名と引数の検証 | 存在しない機能や不適切な入力を実行しないため |
| 権限・承認 | モデルの提案が、そのまま外部操作の許可にならないようにするため |
| 実行範囲と資源の上限 | 暴走、長時間実行、過大なメモリ使用を防ぐため |
| 戻り値・例外の記録 | 成功と失敗を区別し、後から追跡するため |
| 再試行・終了条件 | 結果を得られないループを無制限に続けないため |
| 最終値の照合 | 結果を文章にするときの転記や単位の誤りを検出するため |

MCP（Model Context Protocol）は、AIアプリケーションが外部のツールや情報源と**つながるための取り決め**（プロトコル）である
- ホスト、クライアント、サーバーの役割を分ける構成が公開されている

**説明**

MCPは、冒頭の図でいえばレベル3（ハーネス）の接続部分に位置する
- Transformerの内部に新しい機構を足すものでも、計算の正しさを保証するものでもない
- MCPを使わなくても、関数を直接呼べばツール連携は実装できる

> **プロトコルはつなぎ方を定め、ハーネスが処理を管理し、モデルが要求を生成し、ツールが実行する** この役割分担を崩さないことが重要である

MCPの実装と、ハーネスの設計（ループ、権限、検証、記録）は、M-LLM-5 で自作する

## 問20  RAGとは何か  検索はTransformerのどこに入るのか

**答え**

RAG（Retrieval-Augmented Generation：検索拡張生成）は、**質問に関係する資料を外から探してきて、質問と一緒にモデルの入力に入れる**仕組みである
- モデルの重みに入っている知識だけに頼らず、手元の文書を根拠に答えられる

検索はTransformerの中には入らない **入力の手前**に入る
- 検索で選ばれた文章が、文脈（コンテキスト）としてモデルに渡される

**説明**

<img src="https://class.west.sd.keio.ac.jp/dataai/text/llmu-09-rag.png" width=760>

上の図のように、RAGは準備時と質問時に分かれる

- **準備時**: 文書を読み込んで適切な単位に分割し、必要なら埋め込み（ベクトル）を計算して索引を作る 元の文章、文書名、節やページ、更新日時も保持する
- **質問時**: 質問に関係する部分を検索し、必要なら順位をつけ直し、選ばれた文章を質問と一緒にモデルに渡す
  - 意味的な近さによる検索（埋め込み）と、キーワード検索を組み合わせる実装もある

モデルが全文書を毎回読んでいるわけではない 検索で選ばれた範囲だけが渡される（Lewis et al., 2020）

よくある誤解を正しておく

- **検索用の埋め込みと、Transformerの内部表現は同じではない** どちらもベクトルだが、目的、学習方法、次元、使い方が違う
- RAGは「文書のベクトルをモデルが逆変換して文章を復元する」仕組みではない **ベクトルで候補を見つけ、対応する原文を取り出す**仕組みである
- ベクトルが近いことは、正しい根拠であることを保証しない
  - 似た言葉を含む別の版、否定された条件、単位の違う記述が上位に来ることがある（問22で確かめる）
- **Web検索とRAGは、重なるが同じではない** Web検索は情報を取る方法の1つで、RAGは取った情報を生成につなぐ構成の名前である
  - 対象はWebでも、論文でも、社内文書でもよい
- 最初の資料で足りなければ、モデルが追加の検索語を作り、外側が再検索して戻す、という繰り返しもできる
  - このとき、**生成の繰り返しの外側に、検索とモデル要求の繰り返しが加わる** 推論と行動を交互に行う構成は ReAct として研究されている（Yao et al., 2022）

## 問21  RAGを使うと、モデルは新しいことを学習するのか

**答え**

学習しない 検索した資料は**入力に足される**だけで、モデルの重みは1ビットも変わらない

次の質問で資料を渡さなければ、モデルは前回の資料を使えない

「会話中に学んだ」という言い方には2つの意味があり、区別が要る

- **文脈内学習（In-context Learning）**: 入力にルールや例を書くと、その会話の中ではそれに従える 重みは変わらない（Brown et al., 2020）
- **追加学習（ファインチューニング）**: 大量の例で重みそのものを更新する別の処理である

**説明**

RAGを含むシステムの学習で、検索器や生成器を訓練する研究はある
- しかし、運用中に資料を検索して1回回答したことが、そのままモデルの再学習になるわけではない

次のセルで、「文脈にルールを書くと従う」ことと「そのあいだ重みは変わらない」ことを、同時に確かめる

### 確かめる ― 文脈のルールには従うが、重みは変わらない

何を実装しているか

- 重みの一部（最初の層のクエリ射影、最後の層のFFNの出力射影）のハッシュを取る
- 架空の演算 `⊕`（$X\oplus Y=2X+Y$）を、(1) 文脈に定義を書いて、(2) 定義を書かずに尋ねる
- 生成の前後でハッシュが一致することを `assert` で確かめる

実装の肝

- 定義を書いた場合の `3⊕5` の正解は11である 定義がなければ、モデルは定義を知り得ない
- ハッシュの一致は、重みが1ビットも変わっていないことの確認である

結果の読み方

- 同じモデル、同じ重みで、文脈によって答えが変わる 変わるのは入力と作業状態である

観測上の限界

- 1つのルール、1つのモデルの観察である 文脈内学習がいつ成功するかの一般的な結論ではない

````python
# 文脈で新しいルールに従うが、重みは変わらない
def weight_fingerprint(model):
    layers = model.model.layers
    parts = [layers[0].self_attn.q_proj.weight, layers[-1].mlp.down_proj.weight]
    return [hashlib.sha256(p.detach().float().cpu().numpy().tobytes()).hexdigest()[:16] for p in parts]


before = weight_fingerprint(llm)
with_rule = chat_batch(["新しい演算 ⊕ を、X⊕Y = 2×X + Y と定義する。3⊕5 はいくつか。答えの整数だけを示してください。"], 24, prefill="答え: ")[0][0]
without_rule = chat_batch(["3⊕5 はいくつか。答えの整数だけを示してください。"], 24, prefill="答え: ")[0][0]
after = weight_fingerprint(llm)

assert before == after
print("重みのハッシュ（生成の前）:", before)
print("重みのハッシュ（生成の後）:", after, "→ 一致")
print(f"定義を文脈に書いた場合の 3⊕5 → {with_rule!r}（正解は 11）")
print(f"定義を書かなかった場合の 3⊕5 → {without_rule!r}")
````

結果の読み方

- 重みのハッシュは生成の前後で完全に一致する
- 定義を文脈に書くと `3⊕5` は11（正解）と答え、書かなければ8と答える（`⊕` を足し算として扱ったと考えられる）
- **同じ重みで、文脈が違えば答えが変わる** これが文脈内学習であり、学習（重みの更新）ではない 文脈で与えたルールは、その入力のあいだだけ有効である

## 問22  RAGがあれば、計算も正確になるのか

**答え**

ならない

RAGが持ち込むのは**根拠（係数や条件の値）**であって、計算の能力ではない
- 資料から「係数は17と26」を取り出せても、$24\times17+38\times26-75$ を正しく計算し終えたことにはならない

**RAGが根拠を持ち込み、モデルが関係を解釈し、計算ツールが演算し、最後に検証する** それぞれの役割は違う

**検索が誤れば、計算が正確でも答えは誤る** 検索の誤りは、モデルが直せない

**説明**

次のセルで、「資料を使う計算」の問題を5つの条件で解かせる
- 計算の指示（途中を書く）は同じで、**モデルに見せる資料だけ**を変える

| 条件 | モデルに入力する資料 |
| --- | --- |
| G0 資料なし | 何も与えない 問題文は「資料が取得できなければ推測せず、その旨を答える」よう求めている |
| G1 全文入力 | 資料を全件（4〜64件）入力する |
| G2 検索（製品＋改訂） | 製品名と改訂が一致する資料1件だけを渡す |
| G3 検索（製品だけ・上位1件） | 製品名だけで検索し、最初の1件を渡す（改訂を区別しない） |
| G4 検索（製品だけ・上位2件） | 製品名だけで検索し、上位2件（改訂1と2）を渡す |

検索器は文字列の一致で資料を選ぶ簡単なもので、本物の検索器ではない 比べたいのは、**検索の結果が、モデルの手前で何を決めてしまうか**である

### 確かめる ― 検索の誤りは、モデルが直せない

何を実装しているか

- 問題文から製品名と改訂を読み取り、資料を絞り込む簡単な検索器を作り、5つの条件で解かせて正解数を数える
- G0では、「取得できない」と答えた問題数と、推測して数字を出した問題数を分けて数える

実装の肝

- G3は**検索が一部しか正しくない**場合を再現する
  - 同じ製品名の資料が2件（改訂1と2）あるので、改訂を区別せずに最初の1件を渡すと、半分の問題で誤った資料が渡される
- G4は、検索が候補を絞るだけで、どちらが正しいかの選択はモデルに任せる場合である
- 全文入力は入力が長いので、バッチを小さくしている

結果の読み方

- G2とG3の差が、**検索の精度がそのまま回答の精度の上限になる**ことを示す
- G0で推測した問題があれば、それは「でっち上げ」である 指示があっても起き得る

観測上の限界

- 資料は架空の短い文章である 実際の文書は長く、曖昧で、矛盾を含む 各水準5問である

````python
# 簡単な検索器と、5つの条件
def corpus_text(docs):
    return "\n".join(f"[{d['document_id']}] {d['text']}" for d in docs)


def retrieve(problem, docs, mode):
    prod, rev = re.search(r"製品=(\S+?)、改訂=(\d+)", problem).groups()
    if mode == "exact":                      # 製品と改訂の両方が一致する資料
        return [d for d in docs if f"製品={prod}、改訂={rev}。" in d["text"]][:1]
    hits = [d for d in docs if f"製品={prod}、" in d["text"]]       # 製品名だけで絞る（改訂は区別しない）
    return hits[:1] if mode == "product_top1" else hits[:2]


G_CONDS = {"G0資料なし": lambda q: None,
           "G1全文入力": lambda q: corpora[q["id"]],
           "G2検索(正)": lambda q: retrieve(q["problem"], corpora[q["id"]], "exact"),
           "G3検索(上位1)": lambda q: retrieve(q["problem"], corpora[q["id"]], "product_top1"),
           "G4検索(上位2)": lambda q: retrieve(q["problem"], corpora[q["id"]], "product_top2")}
rag_items = by_family["document_grounded_calculation"]
for cname, fn in G_CONDS.items():
    prompts = []
    for q in rag_items:
        docs = fn(q)
        prompts.append(("" if docs is None else "資料:\n" + corpus_text(docs) + "\n\n")
                       + scratch_prompt(q))
    evaluate(cname, rag_items, prompts, 800, answer_line, batch_size=4 if cname == "G1全文入力" else 8)    # 全文入力は長いので、バッチを小さくする
    print(f"           入力の長さ（平均） {np.mean([len(tok(p).input_ids) for p in prompts]):.0f} トークン、最大 {max(len(tok(p).input_ids) for p in prompts)} トークン")

print()
summarize(list(G_CONDS), "document_grounded_calculation")
g0 = [r for r in results if r["cond"] == "G0資料なし"]
print(f"\nG0（資料なし）: 数字を出した（推測） {sum(r['pred'] is not None for r in g0)}/{len(g0)}問、"
      f"数字を出さなかった（拒否など） {sum(r['pred'] is None for r in g0)}/{len(g0)}問、正解 {sum(r['ok'] for r in g0)}問")
````

結果の読み方

各水準5問、合計20問で、次のようになる

- **G0 資料なし**: 20問すべてで、推測せず「該当資料が取得できない」と答える
  - 正解は0問、推測して数字を出した問題も0問である
  - 問題文が「推測せず、その旨を回答する」よう求めているので、このモデルは指示に従った 常にそうなるとは限らない
- **G1 全文入力**: 全水準で5/5である 入力は平均1960トークン、最大4336トークンで、64件のときは長い
- **G2 検索（製品＋改訂）**: 全水準で5/5である 正答はG1と同じで、入力は平均187トークン（約10分の1）である
- **G3 検索（製品だけ・上位1件）**: 全水準で2/5（合計8/20）である
  - 内訳を見ると、**対象の改訂が1の8問は全問正解、改訂が2の12問は全問不正解**である 検索が改訂を区別せず常に改訂1を渡したので、改訂2の問題には誤った資料が渡された
  - モデルは渡された資料に忠実に計算したので、答えは「整合した誤答」になった
- **G4 検索（製品だけ・上位2件）**: 全水準で5/5である 改訂1と2の両方を渡せば、モデルが問題の指定に合う資料を選べる

ここから言えること

- **検索の精度が、回答の精度の上限になる** モデルがどれほど正確に計算しても、誤った資料からは正しい答えは出ない
  - G3の誤りはモデルの能力の限界ではなく、検索の設計の誤りである
- 候補を1件に絞り切らず、複数件を渡してモデルに選ばせる設計（G4）は、誤りを減らし得る
  - ただし入力が長くなり、関係のない資料に引きずられる危険も増える
- 全文入力（G1）はこの規模では十分に動く 資料が数百件、数万件になれば入力に収まらず、検索が必要になる

観測上の限界

- 資料は架空の短い文章で、製品名と改訂の一致という単純な規則で選べる
  - 実際の文書では曖昧さ、矛盾、長さが問題になる
  - 検索器は文字列の一致で、ベクトル検索やBM25の誤りの出方とは違う

## 問23  検索した文書に命令が書いてあったら、従ってよいのか

**答え**

従ってはいけない

検索で得た文書の中身は**処理対象のデータ**であり、その文書に書かれた命令を実行する権限ではない

外部の資料に「他のファイルを送信せよ」と書いてあっても、それを利用者や管理者からの指示と同じ扱いにしない設計が必要である
- これは**外部の入力と実行の権限を分ける**原則で、RAGやWeb検索を付ければ自動的に安全になるわけではない

**説明**

この攻撃を「間接プロンプトインジェクション」と呼ぶ 仕組みと防御（権限の最小化、危険な操作の承認、信頼できない入力を読んだ後に権限を下げる）については別途学んだ

また、実務では、回答のどの主張がどの資料のどの部分に支えられているかを残す必要がある 引用が付いているだけでは、引用先が実際にその主張を支持しているとは限らない

外部に資料があることと、それが適切なタイミングでモデルに与えられることのあいだには、検索・選択・文脈の編集という処理がある

## 問25  計算機が正確なら、システム全体も正確になるのか

**答え**

ならない 計算ツールが保証するのは、**渡された式を正しく計算する区間**だけである

その手前（資料の取り違え、単位の誤解、誤った立式、誤った引数）と、その後（結果の転記ミス）は保証しない

だから、**モデルの答えとは独立に、外側から確かめる検証**が要る 段階を増やすほど誤りの機会も増えるので、検証がないまま反復を増やしても、全体の正確さは上がらない

**説明**

<img src="https://class.west.sd.keio.ac.jp/dataai/text/llmu-11-integrated.png" width=760>

上の図は、問18の課題（係数を資料から取り、$24A+38B-75$ を求める）を、システム全体として追跡したものである
- 利用者に見える答えは `1321` の4文字だが、裏では取得、立式、演算、検証、表示が分業している 各段階に、確かめるべきことがある

| 段階 | 実行主体 | 確認点 |
| --- | --- | --- |
| 資料取得 | 検索系・ハーネス | 正しい文書・版・対象・単位か |
| 解釈・立式 | LLM | 補正値を一度引くのか、各項で引くのか |
| 演算 | 計算ツール | 式が正しく渡され、実行が成功したか |
| 検証 | ハーネス・検証器 | 単位、適用条件、転記に誤りがないか |
| 回答 | モデルまたは定型の整形 | 検証済みの値を変えていないか |

誤りが累積する様子は、簡単な計算で見える 各段階の正答確率が0.99で独立だと仮定すると、100段階すべてが正しい確率は

$$
0.99^{100}\approx0.366
$$

である これはLLMの実測値ではなく、独立を仮定した計算例である
- 実際の誤りは依存し、途中で直ることもある それでも、**反復を増やすだけでは全体の正確さは保証されない**ことがわかる 検査や修正が効いて初めて、追加の計算が改善につながる

### 確かめる ― 誤りの累積と、検査の効果

何を実装しているか

- 100段階すべてが正しい確率を、式（$0.99^{100}$）と20万回の乱数実験の両方で求める
- 各段階に「誤りの90%を検出して直せる検査」を置いた場合の確率を式で求める

実装の肝

- 乱数実験は「各段階が独立に確率0.99で正しい」という仮定をそのまま実行したもので、式の値と0.005以内で一致することを `assert` で確かめる
- 検査を置くと、1段階で誤りが残る確率は $0.01\times(1-0.9)=0.001$ になる

結果の読み方

- 検査なしと検査ありで、100段階すべてが正しい確率がどれだけ変わるかを読む

観測上の限界

- 独立性を仮定した計算例である
  - 検査そのものが誤る場合は考えていない

````python
# 誤りの累積: 100段階すべてが正しい確率（式と乱数実験）、および検査で誤りの90%を直せる場合
n_steps, p_ok, trials = 100, 0.99, 200_000
rng = np.random.default_rng(0)
exact = p_ok ** n_steps
mc = (rng.random((trials, n_steps)) < p_ok).all(axis=1).mean()
assert abs(exact - mc) < 0.005

p_fixed = 1 - (1 - p_ok) * (1 - 0.9)                          # 誤りの90%を検出して直せる検査を、各段階に置く
exact_fixed = p_fixed ** n_steps
print(f"検査なし: 式 {exact:.3f} / 乱数実験 {mc:.3f}")
print(f"検査あり（誤りの90%を直す）: 1段階の正答確率 {p_fixed:.3f} → 100段階すべて正しい確率 {exact_fixed:.3f}")
````

結果の読み方

- 式の値0.366と、20万回の乱数実験0.368が一致する
- 誤りの90%を直せる検査を各段階に置くと、1段階の正答確率は0.999になり、100段階すべてが正しい確率は0.905に上がる
- 同じ100段階でも、**検査があるかどうか**で全体の正確さは0.37から0.91に変わる 反復を増やすことと、検査・修正を置くことは、セットで考える

### 確かめる ― 検証は、計算の外から確かめる（故障を注入する）

何を実装しているか

- 実行記録から、モデルの答えとは独立に、「依頼どおりの製品の資料を使ったか」「資料の係数から再計算した答えと一致するか」を確かめる検証関数 `verify_run` を作る
- 3つの場合に適用する  **正常**（問18の実行）、**故障1**（検索ツールが依頼と違う製品の資料を返す）、**故障2**（計算ツールが正しい値より1大きい値を返す）

実装の肝

- 故障1では、モデルは与えられた資料に忠実に整合した計算をする
  - だから、**答えだけを見ても誤りは見つからない** 資料の製品名が依頼と合っているかを外側が照合して初めて検出できる
- 故障2では、モデルの式は正しくてもツールの戻り値が誤っている
  - 資料の係数から独立に再計算して照合すれば検出できる

結果の読み方

- 2つの故障では、どちらも回答は文章として整っている
  - それでも検証関数は、異なる理由で誤りを検出する

観測上の限界

- 検証関数が見る範囲（製品名の一致、再計算）の誤りだけを検出する
  - 単位の取り違えや適用条件の誤りは検出できない

````python
# 検証: 実行記録を、モデルの答えとは独立に照合する
def verify_run(product, records, final, n_a=24, n_b=38, correction=75):
    problems = []
    spec = next((r for r in records if r["tool"] == "lookup_spec" and r["ok"]), None)
    if spec is None:
        return ["資料を取得していない"]
    if f"製品={product}、" not in spec["result"]:
        problems.append("取得した資料の製品が、依頼と違う")
    a, b = map(int, re.search(r"係数A=(\d+)、係数B=(\d+)", spec["result"]).groups())
    expected = n_a * a + n_b * b - correction
    if final.strip().replace(",", "") != str(expected):
        problems.append(f"資料の係数から再計算した値 {expected} と、回答 {final!r} が一致しない")
    return problems


# 正常
assert verify_run("装置-007", records, final) == []
print("正常:", final, "→ 検証OK")

# 故障1: 検索ツールが、依頼と違う製品の資料を返す
final1, rec1, _ = run_harness(backend, SPEC_REQUEST, SPEC_TOOLS, {**SPEC_IMPLS, "lookup_spec": lambda product: SPECS["装置-008"]},
                              system=SPEC_SYSTEM)
p1 = verify_run("装置-007", rec1, final1)
print("故障1（資料の取り違え）: 回答", repr(final1), "→ 検証:", p1)
assert p1

# 故障2: 計算ツールが、正しい値より1大きい値を返す
final2, rec2, _ = run_harness(backend, SPEC_REQUEST, SPEC_TOOLS,
                              {**SPEC_IMPLS, "integer_expression": lambda expression: integer_expression(expression) + 1},
                              system=SPEC_SYSTEM)
p2 = verify_run("装置-007", rec2, final2)
print("故障2（計算ツールの不具合）: 回答", repr(final2), "→ 検証:", p2)
assert p2
````

結果の読み方

- **正常**な場合は検証が通る（回答 `1321`）
- **故障1（資料の取り違え）**では、回答は `1125` である これは装置-008の係数（31と12）で `24 * 31 + 38 * 12 - 75` を計算した値で、与えられた資料に対しては整合した答えである 回答の文章だけを見ても誤りは見つからない 検証関数が、取得した資料の製品名が依頼と違うことを検出する
- **故障2（計算ツールの不具合）**では、回答は `1322` である 検証関数が、資料の係数から再計算した値（1321）と回答が一致しないことを検出する
- どちらの故障でも、モデルの回答は文章として整っている 誤りを見つけたのは、モデルの外側の独立した照合である

## 問26  どの部分をLLMに任せ、どの部分を任せないべきか

**答え**

LLMに任せるのは、**自然言語の曖昧さを解釈する部分と、計画を立てる部分**である 確実な演算、根拠の保持、権限を伴う操作、最終値の照合は、明示的なプログラムに任せる

- 求める出力が数値だけなら、検証済みの戻り値を定型処理で表示するほうが、言い換えによる転記の誤りを避けやすい
- 課題の形式が完全に決まっていれば、立式まで通常のプログラムでできることもある

> **LLMをすべての処理の実行者にする必要はない** 解釈・計画と、検索・数値計算・検証を分業させられることが、システムとしての強みである

**説明**

「数値演算をする専用の部品を足せばよいのでは」という考えは合理的で、プログラム実行を使う方式が研究されている（問12の条件F） ただし、どの問題をその部品に渡すか、式や条件をどう組むか、失敗をどう検出するかは残る モデル内の学習済みの「専門家」（問33）と、外部の厳密な計算手続きは、どちらも「専門化」していても性質が違う

100万手順の演算を外部に移せば、Transformerの層数でその100万手順を説明する必要はなくなる しかし、モデルには**正しいプログラムを選ぶ、正しい引数を渡す、結果を正しく解釈する**課題が残る 難しさが消えたのではなく、学習モデルが得意な部分と、明示的な実行・検証が得意な部分に分解したのである

# 学習と小型モデル

## 問27  学習で何を獲得し、使うときに何を実行しているのか

**答え**

学習で獲得するのは**重み**、つまり「入力をどう変換するか」の規則である 使うとき（推論時）に実行するのは、その規則を入力に当てはめる計算と、生成の繰り返しである

Transformerという構造があるだけでは、数学もプログラム生成もツール選択も正確にはならない 構造がある計算を**表現できる**ことと、その重みを学習で**獲得できた**ことは別である

**説明**

| 学習・調整の段階 | 主に与えるもの | 本テキストとの関係 |
| --- | --- | --- |
| 事前学習 | 大量の文章・コードに含まれる構造 | 次トークン予測の基礎的な変換を獲得する |
| SFT（教師あり追加学習） | 指示への応答や解法の例 | 指示の形式、途中結果、ツール要求を学習できる |
| RL（強化学習） | 結果に対する報酬 | 成功につながる生成・問題解決の行動を強化する |
| 蒸留 | 他のモデルの出力や振る舞い | 小さいモデルへ有用な応答・解法の傾向を移す |

これは代表的な整理で、すべてのモデルが同じ順序・方法で訓練されるわけではない

推論モデルは、「層を深く足したモデル」ではない 推論用の中間生成をうまく使うように訓練すれば、層数を極端に増やさなくても、回答の前により多くの計算ができる DeepSeek-R1の研究では、強化学習による推論能力の獲得と、小型モデルへの蒸留が示されている（DeepSeek-AI, 2025） ここでも、学習が高品質な手順の生成を促すことと、推論時に計算時間を払うことは別で、学習で長い計算が消えるわけではない

## 問28  小さなローカルモデルでも推論できるのは、なぜか

**答え**

有限の規則や解法の傾向は、小型モデルにも学習され得るからである さらに、小型モデルも生成のたびに繰り返し使える 必要なら、同じコンピュータの中の計算ツールや検索も使える

だから、小型モデルが問題を解けたとき、「裏で巨大なモデルが動いているはずだ」と考える必要はない 公開されている小型の推論モデルや、思考モードを持つモデルは実在する

**説明**

「小さなモデルが解けたなら、巨大な表現空間は要らないのか」という疑問には、こう答える その課題には巨大モデルが必須ではなかった、とは言える しかし、その小さなモデルがどの規則を学習し、どれだけ生成し、何を外部に委ねたかは、別途確かめる必要がある

「ローカルで動いている」という表示だけでは、アプリケーションが外部通信や別プロセスを使っていないことまでは証明できない 実際の処理を知るには、実行の設定と記録を見る（問42）

## 問29  学習は並列にできるのに、生成はなぜ1つずつなのか

**答え**

学習では、正解のトークン列が最初から全部わかっている だから「ある位置までの入力」を全位置ぶん同時に用意でき、因果マスクのもとで多数の位置の予測と損失をまとめて計算できる

生成では、次に何を出すかがまだ決まっていない 選んだトークンを次の入力に使う、という依存があるので、1つずつ進めるしかない **学習が並列化できることと、生成が任意に並列化できることは別である**

**説明**

問6のPrefillは、入力部分については学習と同じ並列処理をしている 違いが出るのはDecodeで、ここは生成したトークンへの依存がある 投機的デコーディング（問36）は、この依存を保ったまま高速化する工夫である

# 実行基盤の共通化

## 問30  Ollamaなどは、なぜ複雑な周辺機能まで共通の仕組みで扱えるのか

**答え**

Ollamaのような実行基盤が共通化しているのは、**モデルを実行する仕組み、トークンを1つずつ生成する繰り返し、思考やツール呼び出しを受け渡す入出力の形式**である 複雑な問題解決の仕組み全体をモデルファイルから読み取っているのではない

- モデルが何を推論し、どのツールを選ぶかという**能力**は、学習済みの重みに依存する
- ツールの実行、検索、検証といった**周辺の処理**は、別のプログラム（ハーネス、問18）が担当する

つまり、共通化が成り立つ理由は、複雑な処理をすべてモデルの中に表現しているからではなく、**モデルと周辺のあいだに共通のインターフェースを置き、役割を分離しているから**である

**説明**

<img src="https://class.west.sd.keio.ac.jp/dataai/text/llmu-13-layers.png" width=760>

上の図のように、Ollamaを使うシステムは6つの層に分けられる（実装上は複数の層が同じソフトウェアに含まれることもある）

| 層 | 内容 | 共通化の程度 |
| --- | --- | --- |
| 1 学習済みパラメータ | 重み | モデルごとに異なる（ファイル形式は共通化できる） |
| 2 モデル定義 | アーキテクチャ名、層数、次元、トークナイザ、チャットの書式 | モデルごとに異なる（記述の形式は共通化できる） |
| 3 推論ランタイム | モデルの実行、生成の繰り返し、KVキャッシュ、抽選 | 実行基盤が共通化している ただし対応する構造に限る |
| 4 API | メッセージ、思考、ツール呼び出しの共通の入出力形式 | 実行基盤が共通化している |
| 5 ハーネス | ツール実行、RAG、処理の繰り返し、停止条件、検証 | 外部のアプリケーションが実装する部分が大きい |
| 6 アプリケーション | 画面、履歴、結果の表示 | 外部のアプリケーションが実装する |

層3の「生成の繰り返し」がモデルに依存しないことは、問10の擬似コードからわかる モデルごとに違うのは `model.predict_next` の中身（アーキテクチャごとの計算グラフ）だけで、トークンを選ぶ、追加する、終了を判定する、キャッシュを管理する部分は共通に書ける ただし、そのモデルの構造に**対応している**ことが前提である

層4では、実行基盤がモデルごとの出力の違いを吸収する Ollamaのチャット応答は、推論を `message.thinking`、最終回答を `message.content` に分けて返す 推論を有効にするかは、リクエストの `think` で指定する（`true`、`false`、モデルの既定、gpt-oss系では `low`、`medium`、`high`） 指定できる値はモデルによって違い、`/api/show` で確かめる

```json
{"message": {"role": "assistant", "thinking": "まず数値を確認する必要がある", "content": "計算結果は123である"}}
```

これは、問13で作った `split_message` と同じ発想である モデルごとの特殊トークン（Qwen3の `</think>` など）を、共通のフィールドに変換している

ツール呼び出しも同じである Ollamaの公式文書では、次の手順が説明されている

1. リクエストに `tools`（`type: "function"`、`name`、`description`、`parameters`）を付けて送る
2. モデルがツールを必要と判断すると、応答の `message.tool_calls` に関数名と引数が入る
3. **アプリケーション側のコードが**その関数を実行する モデルは実行しない
4. 結果を `role: "tool"`（`tool_name`、`content`）のメッセージとして履歴に追加し、再び送る
5. モデルが結果を使って回答を作る `tool_calls` がなくなるまで繰り返す

標準化されているのは、「**どの関数を、どんな引数で呼びたいか**」という情報の表現である 関数の実装や、実行してよいかの権限は、モデルも実行基盤も持たない

Web検索も同じ構図である Ollamaには Web検索のAPIがあるが、これはOllamaが提供する外部のホスト型サービスで、アカウントとAPIキーが必要である ローカルのTransformerの演算ではない 公式のエージェント例では、`while True` のループでモデルを呼び、`tool_calls` があれば対応する関数（`web_search`、`web_fetch`）を実行して結果を `tool` のメッセージとして追加し、`tool_calls` がなくなれば `break` する 繰り返し、ツール実行、終了判定を担当しているのはPythonのプログラムである

## 問31  モデルファイルには、何が入っているのか

**答え**

入っているのは、**重み（テンソル）**と、**メタデータ**（アーキテクチャの名前、層数、次元、トークナイザ、チャットの書式）である 計算の手順（計算グラフ）は入っていない 手順は、実行エンジンが「アーキテクチャの名前」を見て、自分が持っている実装の中から選ぶ

だから、PyTorchの重みを保存しただけでは、どんなモデルでも動くわけではない **ファイルの形式が共通であること**と、**計算の中身が共通であること**は違う

**説明**

llama.cpp系の実行基盤が使う GGUF 形式には、テンソルのほかに次のメタデータが入る（GGUF仕様）

- `general.architecture`: このモデルが実装するアーキテクチャの名前（必須）
- `[アーキテクチャ名].block_count`、`.embedding_length`、`.context_length`、`.attention.head_count`: 層数、次元、文脈長、ヘッド数
- `tokenizer.ggml.*`: トークンの一覧、結合規則、特殊トークンの番号
- `tokenizer.chat_template`: モデルが期待する入力形式を示すテンプレート

未対応のアーキテクチャは、ファイルを変換しただけでは動かない llama.cppの開発資料は、新しいアーキテクチャを追加するには、GGUFへの変換処理、テンソル名の対応表、そして実行エンジン側の**計算グラフの実装**が必要だと説明している GGUFの仕様書も、計算グラフをファイルに含める拡張は「今後の検討事項」としている

Ollamaには Modelfile というモデル設定の形式もある

```text
FROM llama3.2
PARAMETER temperature 0.2
PARAMETER num_ctx 8192
SYSTEM You are a helpful research assistant.
```

- `FROM`（必須）: 元になるモデル 既存のモデル、Safetensorsのディレクトリ、GGUFファイルを指定できる
- `PARAMETER`: `temperature`、`num_ctx`、`seed`、`stop` など、動かすときの設定
- `TEMPLATE`: モデルごとのチャットの書式
- `SYSTEM`、`MESSAGE`: システムメッセージ、例として与える会話

ここには、複雑なRAGの手順や、プログラム実行の手順の全体は含まれない 保存する対象は3つに分けて考える

| 保存対象 | 内容 | 共通化の程度 |
| --- | --- | --- |
| モデル定義 | アーキテクチャ、重み、トークナイザ、チャットの書式 | GGUFやSafetensors、Modelfileで、かなり共通化できる |
| 実行設定 | 生成方法、推論モード、文脈長、使えるツールの宣言 | 実行基盤のAPIで、かなり共通化できる |
| ワークフロー定義 | 何を検索し、どのプログラムを実行し、結果をどう検証し、いつ終了するか | アプリケーションやエージェント基盤の担当 |

共通のAPIやツール接続の取り決め（MCP、問19）は使えるが、**任意のワークフローをモデルファイルだけで記述できるわけではない**

### 確かめる ― 同じ会話が、モデル系列ごとに別の文字列になる

何を実装しているか

- 同じ会話（システムメッセージ＋利用者の発言）を、4つのモデルのチャットテンプレートで文字列にして比べる
- 文字列が3種類に分かれる（2つのモデルは同じ書式を使う）ことを `assert` で確かめる

実装の肝

- `apply_chat_template` は、トークナイザに付属するテンプレートで、会話を**そのモデルが学習した書式**の1本の文字列に直す
- 実行基盤が共通の入力形式（`messages` のリスト）を受け付けられるのは、モデルごとの違いをこのテンプレートの層で吸収しているからである

結果の読み方

- 特殊トークンの名前、役割の書き方、区切りが系列ごとに違う QwenとSmolLM2は同じ書式（ChatML）を使うので、文字列が一致する
- 利用者に見える `messages` は同じでも、モデルに入る文字列は系列によって違う 書式を間違えると、モデルは学習した形式で入力を受け取れず、性能が落ちる

観測上の限界

- 比べるのはトークナイザのファイル（数MB）だけで、モデルの重みは読み込まない

````python
# 同じ会話を、4つのモデルのチャットテンプレートで文字列にする
messages = [{"role": "system", "content": "あなたは丁寧な助手である"},
            {"role": "user", "content": "1+1は？"}]
TEMPLATE_MODELS = ["Qwen/Qwen3-4B-Instruct-2507", "HuggingFaceTB/SmolLM2-135M-Instruct",
                   "microsoft/Phi-3-mini-4k-instruct", "TinyLlama/TinyLlama-1.1B-Chat-v1.0"]
rendered = {}
for name in TEMPLATE_MODELS:
    t = AutoTokenizer.from_pretrained(name)
    rendered[name] = t.apply_chat_template(messages, tokenize=False, add_generation_prompt=True)
    print(f"--- {name}")
    print(rendered[name])

# QwenとSmolLM2は、同じ書式（ChatML）を使う 他の2つは、それぞれ別の書式である
assert rendered[TEMPLATE_MODELS[0]] == rendered[TEMPLATE_MODELS[1]]
assert len(set(rendered.values())) == 3
print("\n4つのモデルの入力形式は、3種類に分かれた（QwenとSmolLM2は共通のChatML形式）")
````

### 確かめる ― GGUFファイルを自分で書いて読む

何を実装しているか

- `gguf` ライブラリで、ごく小さな架空のGGUFファイルを書き出し、読み戻して中身を表示する

実装の肝

- 形式の見本であり、重みは乱数で、言語モデルとしては動かない
- ファイルに入っているのは「メタデータ」と「テンソル」である 計算の手順は入っていない

結果の読み方

- `general.architecture` に書かれた名前を実行エンジンが知らなければ、このファイルは実行できない これが「形式が共通でも、アーキテクチャごとの実装は共通ではない」ことの意味である

観測上の限界

- 実際のモデルのGGUFは、量子化された大きなテンソルを多数含む ここでは構造だけを見る

````python
# 架空の小さなGGUFを書き出し、読み戻す（形式の見本であり、動作する言語モデルではない）
import os
import tempfile

import gguf

path = os.path.join(tempfile.mkdtemp(), "toy.gguf")
writer = gguf.GGUFWriter(path, "llama")                        # general.architecture = "llama"
writer.add_block_count(2)                                      # llama.block_count
writer.add_embedding_length(8)                                 # llama.embedding_length
writer.add_context_length(128)
writer.add_head_count(2)
writer.add_tokenizer_model("llama")                            # tokenizer.ggml.model
writer.add_token_list(["<unk>", "<s>", "</s>", "a", "b"])      # tokenizer.ggml.tokens
writer.add_chat_template("{{ messages }}")                     # tokenizer.chat_template（見本）
writer.add_tensor("token_embd.weight", np.random.default_rng(0).normal(size=(5, 8)).astype(np.float32))
writer.write_header_to_file()
writer.write_kv_data_to_file()
writer.write_tensors_to_file()
writer.close()

reader = gguf.GGUFReader(path)


def field_value(key):
    f = reader.fields[key]
    return bytes(f.parts[f.data[0]]).decode() if f.types[0] == gguf.GGUFValueType.STRING else int(f.parts[f.data[0]][0])


print(f"ファイルの大きさ: {os.path.getsize(path)} バイト")
print("メタデータ:")
for key in reader.fields:
    if key.startswith(("general.arch", "llama.", "tokenizer.chat")):
        print(f"  {key} = {field_value(key)!r}")
print("テンソル:", [(t.name, tuple(int(x) for x in t.shape)) for t in reader.tensors])
assert field_value("general.architecture") == "llama" and field_value("llama.block_count") == 2
assert [t.name for t in reader.tensors] == ["token_embd.weight"]
````

## 問32  モデルを入れ替えても、同じように動くのか

**答え**

**インターフェースは共通でも、動作や能力は共通ではない** 同じハーネスにモデルを差し替えれば「動かす」ことはできるが、同じ品質で動く保証はない

| 機能 | 共通化できる部分 | モデル依存・外部実装が必要な部分 |
| --- | --- | --- |
| 次トークン生成 | 実行ループ | モデルの構造、重み |
| 思考過程 | 出力のフィールド（`thinking` と `content`） | 推論の能力、推論トークンの形式 |
| ツール呼び出し | 関数名と引数の表現 | ツールを選ぶ能力、実際の関数の実行 |
| 数学計算 | 計算ツールのインターフェース | 計算を委ねる判断、計算プログラム |
| RAG | 埋め込み生成のAPI | 文書の分割、検索データベース、検索戦略、引用の管理 |
| Web検索 | 検索要求と結果の受け渡し | 検索サービス、ネットワーク、結果の評価 |
| 複雑な処理手順 | 呼び出しと結果の繰り返し | 実行制御、停止条件、例外処理、状態管理 |

> AIシステム ＝ モデル ＋ 推論ランタイム ＋ 実行制御 ＋ 外部機能

各要素の境界を共通のインターフェースで接続できるので、モデルを取り替えても周辺のプログラムを作り直さずに動かせる ただし、同じインターフェースに対応しているだけでは、同じ推論能力や同じツール選択の品質は保証されない 「モデルファイルの可搬性」「推論ランタイムの共通化」「ハーネスの可搬性」は、別々の概念である

**説明**

次のセルで、2つのことを分けて確かめる

1. ハーネスのコードを**変えずに**、裏のモデル（バックエンド）を取り替えられるか（可搬性）
2. 取り替えたモデルが、**同じ品質で**動くか（能力）

### 確かめる ― 偽のバックエンドで、ハーネスだけを動かす

何を実装しているか

- **台本どおりに応答する偽のバックエンド**（`ScriptedBackend`）を作り、問18の `run_harness` を一切変更せずに動かす

実装の肝

- 偽のバックエンドは、モデルを使わず、決まった要求（`lookup_spec`、`integer_expression`、最終回答）を順に返す これでハーネス側の動作だけを決定的にテストできる
- ハーネスは、モデルが何者かを知らない `chat()` の返り値の形だけに依存する

結果の読み方

- 偽のバックエンドで `1321` が返り、実行記録に2つのツールが順に残れば、ハーネスの可搬性は確認できる

観測上の限界

- 台本は「正しい手順」の1通りだけである 誤った要求を返す台本で、ハーネスの例外処理を試すのは課題3とする

````python
# 台本どおりに応答する偽のバックエンド（ハーネス側だけを決定的にテストするため）
class ScriptedBackend:
    def __init__(self, script):
        self.script, self.i = script, 0

    def chat(self, messages, tools=None, think=None):
        msg = self.script[self.i]
        self.i += 1
        return {"model": "scripted", "message": {"role": "assistant", "content": "", "thinking": "", "tool_calls": [], **msg}}


def call(name, **args):
    return {"tool_calls": [{"function": {"name": name, "arguments": args}}]}


script = [call("lookup_spec", product="装置-007"),
          call("integer_expression", expression="24 * 17 + 38 * 26 - 75"),
          {"content": "1321"}]
final, records, _ = run_harness(ScriptedBackend(script), SPEC_REQUEST, SPEC_TOOLS, SPEC_IMPLS)
assert final == "1321" and [r["tool"] for r in records] == ["lookup_spec", "integer_expression"] and all(r["ok"] for r in records)
print("偽のバックエンド:", final, "/ 実行記録:", [(r["tool"], r["result"][:40]) for r in records])
````

- 1つの課題と2つのモデルだけである 小型モデルが常にこの失敗をするとは言えない

## 問33  MoEの「専門家」は、電卓のような専用部品か

**答え**

違う MoE（Mixture of Experts：専門家混合）は、入力に応じて複数の小さなネットワークのうち一部だけを使う構成である（公開例にDeepSeek-V3がある） 「専門家」は学習されたニューラルネットワークであって、「数学の専門家」や、厳密な演算命令を実行する電卓が内部に並んでいるわけではない

特定の専門家が計算に寄与することはあり得るが、**MoEだから厳密な算術ができる、任意回数の反復ができる**とは言えない モデルの中で専門家を選ぶことと、外部のツールを選ぶこと（問18）は、別の仕組みである

**説明**

LLMの中身がTransformerだけとも限らない Jamba は、TransformerとMamba（状態空間モデル）を組み合わせた公開例である（Lieber et al., 2024） LLMという分類を、特定のTransformerの内部図と同一視してはいけない

## 問34  テキストを出さずに、内部だけで反復する方式はあるのか

**答え**

ある 通常の生成は「トークンを1つ選んで入力に戻す」が、**潜在反復**では、連続値の内部表現（隠れ状態）をトークンにせず、そのまま次の処理に戻す Coconut（Chain of Continuous Thought）は、最後の隠れ状態を次の入力埋め込みに直接戻す方式である（Hao et al., 2024） 同じブロックを再帰的に使って推論時の深さを増やす研究もある（Geiping et al., 2025）

ただし、これは**特別な構造と訓練が要る**方式で、標準のTransformerに推論時だけ同じ処理を繰り返し足せば正しくなるわけではない

**説明**

<img src="https://class.west.sd.keio.ac.jp/dataai/text/llmu-12-loops.png" width=760>

上の図のように、「何度も考える」には戻す場所の違う3種類がある 概念的には

$$
h_{k+1}=F_\theta(h_k,x)
$$

という状態更新を繰り返す（各方式の厳密な実装が同じだという意味ではない） **「非表示の推論トークン」（問13）と「トークンにしない潜在状態の反復」は別物である** どちらも画面には途中が見えないが、計算グラフが違う

関連して、意味のないトークンでも計算に役立つ場合がある 専用の学習条件では、`.` のような埋め草トークン（filler token）を足すことで、即答では難しい課題を処理できる事例が報告されている（Pfau et al., 2024） 同じ文字のトークンでも、その位置の内部表現は文脈に依存するので、**見えているトークンの意味だけでは、内部で使われた計算を説明しきれない** ただし、学習は容易ではなく特定の教師信号が必要だった この結果を「点を1000個付ければ1000手順の任意の計算ができる」と一般化してはいけない

## 問35  答えの候補を複数作って多数決すれば、正しくなるのか

**答え**

必ずではない Self-Consistency（自己整合性）は、複数の推論経路を抽選で生成し、一致した答えを選ぶ方法である（Wang et al., 2022） Tree of Thoughts は、途中の候補を探索して評価し、必要なら戻る（Yao et al., 2023）

これらで増えているのは、1つのモデルの層数ではなく、**候補の生成と評価を含むシステム全体の計算量**である そして、**多数決は証明ではない** 候補が同じ誤りを共有していれば、誤答が多数を占める 検証器が不正確なら、候補を増やしても選択の誤りは残る

**説明**

次のコードで、16桁の加算5問について、温度0.8で10本の候補を作り、多数決で答えを選ぶ 「貪欲法の1本」「抽選した1本の平均」「10本の多数決」の正答率を比べる 候補が互いに独立に誤るなら多数決は効くが、同じ種類の誤りを共有すると効かない

実行に時間がかかるので、**実行したい場合のみ**試すこと 結果は次の「この実行結果は次の通り」に示す

    return [answer_line(tok.decode(row[enc["input_ids"].shape[1]:], skip_special_tokens=True)) for row in out]

print(f"貪欲法の1本 {greedy_ok:.2f} / サンプリング1本の平均 {np.mean(single):.2f} / 10本の多数決 {np.mean(vote_ok):.2f}")

## 問36  投機的デコーディングで速くなるなら、計算も増えるのか

**答え**

増えない 投機的デコーディング（speculative decoding）は、軽い補助モデルが候補の列を提案し、対象モデルがまとめて検証する方式である 条件を満たす実装では、**対象モデルの出力分布を変えずに**生成を速くできる（Leviathan et al., 2023、Chen et al., 2023）

速くなるのは「同じ答えを出すまでの時間」であって、解ける問題が増えるわけではない 「1回の前向き計算＝1トークン」という数え方に例外ができるだけで、候補の作成と検証の費用は存在する **生成が速くなったことから、長い依存を費用なしで解けるとは結論できない**

**説明**

次のセルで、補助モデル（0.6B）の提案を対象モデル（4B）が検証する生成と、通常の貪欲法を同じ入力で実行し、出力が一致するかと所要時間を比べる 補助モデルと対象モデルは同じトークナイザでなければならない

### 確かめる ― 補助モデルを使っても、出力は変わらないか

何を実装しているか

- Hugging Faceの `assistant_model` で、補助モデルの候補を対象モデルが検証する生成を行い、通常の貪欲法と出力を比べる

実装の肝

- 補助モデルは候補を提案するだけで、採用するかは対象モデルが決める 貪欲法では、対象モデルが自分で選んだはずのトークンだけが採用されるので、理論上は出力が一致する

結果の読み方

- 出力が一致すれば、「分布を保ったまま高速化できる」ことの確認になる 低精度の計算では、まとめて検証したときの丸めが変わり、まれに一致しないことがある
- 所要時間は、補助モデルの提案が採用される割合に依存する 提案が外れやすい課題では速くならないこともある

観測上の限界

- 1つの入力、1回の測定である

````python
# 投機的デコーディング: 補助モデル（0.6B）の提案を、対象モデル（4B）が検証する
q = next(q for q in by_family["conditional_state_updates"] if q["level"] == 4)
text = tok.apply_chat_template([{"role": "user", "content": scratch_prompt(q)}],
                               tokenize=False, add_generation_prompt=True)
enc = tok(text, return_tensors="pt").to(device)
N_SPEC = 200
common = dict(max_new_tokens=N_SPEC, do_sample=False, temperature=None, top_p=None, top_k=None)


def gen(**kw):
    torch.cuda.synchronize()
    t0 = time.time()
    with torch.no_grad():
        out = llm.generate(**enc, **common, **kw)[0, enc["input_ids"].shape[1]:].tolist()
    torch.cuda.synchronize()
    return out, time.time() - t0


gen()                                                                   # ウォームアップ
plain, t_plain = gen()
spec, t_spec = gen(assistant_model=small)
same = plain == spec
first_diff = next((i for i, (a, b) in enumerate(zip(plain, spec)) if a != b), None)
print(f"通常の貪欲法: {t_plain:.2f}秒 / 補助モデルあり: {t_spec:.2f}秒（{t_plain / t_spec:.2f}倍）")
print(f"出力の一致: {same}" + ("" if same else f"（最初の不一致は {first_diff} 番目のトークン、全 {N_SPEC} トークン）"))
````

結果の読み方

- 補助モデルを使った生成と通常の貪欲法で、200トークンの出力が一致する
- 一方、所要時間は補助モデルありのほうが**長い**（通常の0.6〜0.8倍の速さで、実行のたびに値は変わる） 投機的デコーディングは常に速くなるわけではない
- 速くならない理由の候補は、(1) 対象モデル（4B）が小さく1トークンの時間が短いので、提案と検証の費用が相対的に大きい、(2) 補助モデル（0.6B）の提案が採用される割合が低い、である どちらが主因かは、採用率を測っていないので確かめていない

## 問37  画像を理解するLLMは、Transformerが画像を読んでいるのか

**答え**

「画像を理解した」は、画像を扱うモデルを含む**システムの機能**を述べた言い方である 画像ファイルの読み込み、画像をベクトルに変える処理、系列モデルへの接続、外部の処理、表示は、それぞれ別の部品が担当する 文字のトークンだけを扱う標準のTransformerが、画像ファイルの読み込みからすべてを単独で行っているわけではない

# 理論と限界

## 問38  Transformerは、理論的に何ができて、何ができないのか

**答え**

理論の結果は、**条件付き**で読む必要がある 「固定深さのTransformerは、ある種の逐次的な課題に制約がある」という結果と、「中間トークンを生成すれば表現能力が広がる」という結果の両方があり、どちらにも、精度や深さや正規化についての仮定が付いている

主張を4段階に分けると、何が言えて何が言えないかがはっきりする

| 主張 | 必要な根拠 |
| --- | --- |
| その構造で計算を表現できる | 数学的な構成や、表現能力の定理 |
| 学習によってその計算を獲得できた | 学習実験、汎化実験、内部の解析 |
| あるモデルが、ある入力で正答した | 条件を記録した実行結果と、独立した正解 |
| 対象範囲のすべての入力で正しい | 仕様に対する検証・証明、または有限範囲の完全検査 |

上の段階を、次の段階の証拠として自動的に使うことはできない 本テキストの実験が示しているのは、3段目までである

**説明**

- **固定深さの制約**: MerrillとSabharwalは、入力長に対して対数的な数値精度などの条件を置いたTransformerが、定数深さのしきい値回路で模擬できることを示した その結果と計算量クラスの仮定から、ある種の逐次的課題への制約が導かれる（Merrill and Sabharwal, 2022） これは「有限精度だから長い問題は全部解けない」「100層未満なら100手順以上は不可能」という主張ではない 計算量クラスの分離を仮定した制約を、無条件の不可能性として紹介してはいけない
- **中間生成の効果**: 同じ著者らは、中間トークンの生成ステップ数に応じて表現能力がどう変わるかを調べ、多項式個の中間ステップと多項式時間計算の関係を特徴づけた（Merrill and Sabharwal, 2023） これは「時間方向へ計算を展開する」という説明（問6）が比喩ではないことを支持する ただし、特定の実用モデルがその構成を学習済みであることは、別に示す必要がある
- **見かけの難しさから深さは決まらない**: 長い数の乗算は筆算では多数の操作を要するが、それだけで「どんな回路でも同じだけの深さが必要」とは言えない 逆に、出力が1ビットでも複雑な処理が必要な問題は作れる **答えの長さ、入力の長さ、プログラムの行数、並列化後の深さは別の量である**
- **表現空間の大きさだけでは決まらない**: 有限精度の $d$ 次元ベクトルは、各要素が $b$ ビットなら最大 $2^{bd}$ 通りである 無限の情報は入らない 一方で、モデルは位置ごとの状態や外部の記憶も使えるので、1本のベクトルの容量だけでシステム全体の能力を判断するのも不十分である 表現容量は必要な一要素であり、反復、状態の保持、演算方法、学習、入出力、検証を合わせて考える

## 問39  「Turing完全」なら、何でも解けるのか

**答え**

解けない 「Turing完全」は、適切な計算時間と記憶を許せば汎用の計算を模擬できる、という**表現能力**の主張である 有限の重み、有限の精度、有限の文脈、有限の生成予算を持つ実用モデルが、どんな問題でも無制限に正しく解けるという意味ではない

Transformerに関する完全性の議論を読むときも、反復、正規化、精度、記憶についての条件を確かめる

## 問40  ある桁数で正解すれば、もっと長い桁でも解けるのか

**答え**

言えない 訓練より長い入力に対応できること（長さ汎化）には、別の根拠が要る

Position Coupling の研究では、桁の対応などの構造を位置情報に組み込むことで、1〜30桁の加算で学習したモデルを200桁まで汎化させる実験が報告されている（Cho et al., 2024） これは適切な表現と学習が重要であることを示す例だが、200桁での成功は無限桁の保証にはならないし、その構造を持たない一般のモデルにそのまま当てはまるわけでもない

**説明**

「浮動小数点のネットワークで整数の正解を出せるのか」という疑問にも触れておく 有限の入力範囲なら、連続的な内部表現から離散的な正解を区別することに原理的な矛盾はない 内部の値が整数そのものと一致する必要もない ただし、区別の余裕、丸め、量子化、入力長の変化で誤る可能性があり、有限範囲の成功が任意桁の整数計算を保証するわけではない 問12の実験でも、加算は8桁や16桁で誤りが出始めた

## 問41  「正しく計算できる」には、どんな水準があるのか

**答え**

少なくとも5つの水準があり、上の水準の確認は、下の水準の確認を含まない

| 水準 | 確認したこと | まだ確認していないこと |
| --- | --- | --- |
| 例題で正答した | その入力への答えが正しかった | 未知の分布や長い入力での正確さ |
| 多数のテストで高正答率だった | 評価条件で性能が高かった | すべての入力での保証、内部の手順 |
| 手順が妥当だった | 説明された手順を検算できた | 実際の内部計算がその説明どおりだったか |
| 外部計算が正しく実行された | 渡したプログラムや式を実行できた | 元の問題を正しく立式したか |
| 形式的に検証された | 明示した仕様と仮定の範囲で正しさを示せた | 仕様が利用者の意図を完全に表しているか |

この区別があれば、「正しい文字列を出した」という観察を過大に解釈せず、同時にモデルの有用な能力を不当に否定することも避けられる

**説明**

- 外部計算でも、数値の意味と精度は決める必要がある Pythonの整数は任意精度だが、浮動小数点には丸めがあり、十進の小数をすべて厳密に表せるわけではない 整数として厳密に求めるのか、分数か、許容誤差のある数値解かを区別する 適切な型でも、誤った式や単位を渡せば「正しく実行された誤った計算」になる
- 検証の種類は、問題に合わせる 算術なら独立した再計算、プログラムならテスト、文献に基づく回答なら原文との照合、論理なら前提と導出の検査である 「もう一度考えて」と同じモデルに頼むのは再生成の一種で、同じ誤りを繰り返す可能性があり、独立した検証とは違う
- **出典、実行コード、入力、戻り値、検証条件を残すほうが、長い思考の説明だけを示すより、外部から正しさを確かめやすい** モデル単体の正答率を上げることと、システム全体として誤りを検出・抑制することは、並行して考える

## 問42  本当にどこで計算したかは、どう確かめるのか

**答え**

**変える要因を1つずつ分けて、同じ問題で測る** 本テキストの実験は、次の計画に沿って行った

| 条件 | 設定 | 表示 | 本テキストでの実装 | 調べること |
| --- | --- | --- | --- | --- |
| A 直接回答 | 中間生成・ツール・追加ループなし | 最終回答 | A（書き出しを固定）、A0（指示だけ） | モデル内部と回答生成だけの性能 |
| B 中間生成 | 中間結果を生成してから回答 ツールなし | 中間結果と回答 | B | 中間生成の効果 |
| C 表示だけ変更 | Bと同一の生成結果を使う | 最終回答だけ | C | 表示を隠すこと自体は計算を消さない |
| D 別の呼び出しで答えだけにする | Bの応答を、2回目の呼び出しに読ませる | 最終回答だけ | D | 表示を短くする処理をモデルに任せても、計算は消えない 書き写しの誤りの有無 |
| E 推論予算を変更 | 推論モデルの設定や上限を変える | 最終回答だけ | E（推論トークン数の計測） | 非表示の計算と性能の関係 |
| F 外部計算 | モデルが立式し、計算ツールを使う | 最終回答 | F | 演算の外部化と、立式の誤りの切り分け |
| G 検索＋外部計算 | 必要な値を資料から取得して計算 | 出典と回答 | G0〜G4、ハーネスの実行 | 取得、解釈、演算、転記の各誤り |

注意が2つある Cで別の指示を使って生成し直すと、表示以外も変わる 「同じ生成の表示だけ隠す」実験と、「答えだけと指示して生成し直す」実験（A0）、「モデルをもう一度呼んで答えだけにさせる」実験（D）を混同しない Eで異なるモデルを比べると、重みと学習の差が混ざる できる限り同一モデル内の設定差を使い、避けられない差は明記する

**説明**

問題を難しくする軸も分ける **桁数**は長さへの汎化、**途中状態の数**は手順の保持と更新、**分岐や規則の変更**は新しい条件への追従、**必要な資料の数と紛らわしさ**は検索と統合を調べる 公開の例題だけでなく、新しく生成した数字や架空の規則を使うと、既知の解答の再現だけで成功することを減らせる ただし、長い命令列を作っただけで、その長さが理論的に不可避な深さだとは言えない

記録する情報は、次の通りである

| 区分 | 記録内容 |

| 入力 | 完全な指示、チャットテンプレート、トークン化、入力資料 |
| 生成 | 出力の上限、選択方法、乱数の条件、思考の設定 |
| 非表示処理 | 提供される範囲の推論トークン数、追加ループ、要約・再生成 |
| 外部処理 | ツールの定義、呼び出し、引数、戻り値、エラー、通信 |
| 性能 | 正答率、誤りの箇所、文脈長、実行時間、メモリ、生成トークン数 |
| 検証 | 独立した正解の作り方、採点条件、許容誤差 |

商用の画面だけでは判定が難しい 回答までの時間だけでは推論量は決まらない（待ち行列、入力の長さ、通信、ツールの待ち、ハードウェアも関係する） モデル自身に「Pythonを使いましたか」「何ステップ考えましたか」と聞いて得た文章も、計測の代わりにはならない 必要なのは、可能な範囲でのツール履歴、利用量の記録、実装の確認である

提供側が公開していない部分は、**観測できなかった**と記録する 「見えないから存在しない」とも「見えないから高度な機構がある」とも結論しない 非公開の商用モデルについて、層数、ルーター、潜在反復、追加の検証器の有無を、回答の見かけだけから断定することはできない

### 確かめる ― 実験の条件と結果を記録として残す

何を実装しているか

- 実行環境、モデル名、生成の設定、問題集の条件、外部処理の上限、条件ごとの正解数を1つの辞書にまとめ、JSONファイルに保存する

実装の肝

- 条件ごとの正解数は `results` から集計し直す 手で書き写さない
- 観測できなかった項目（`not_observed`）も記録に含める 「見えなかった」ことを「無かった」ことと区別するためである

結果の読み方

- 表示されるJSONが、この実行の記録である 後から結果を比べるときは、この記録が一致しているかを先に確かめる

観測上の限界

- 記録しているのは、本テキストで制御できた範囲である 乱数の状態やGPUの演算の細部までは記録していない

````python
# 実験の記録: 条件、実行環境、結果を1つの辞書にまとめて保存する
agg = defaultdict(lambda: [0, 0, 0])
# resultsが未定義の場合は空リストとして扱い、エラーを防ぐ
active_results = globals().get("results", [])
for r in active_results:
    agg[r["cond"]][0] += r["ok"]
    agg[r["cond"]][1] += 1
    agg[r["cond"]][2] += r["n_tokens"]

# 必要な変数のフォールバック処理
active_dtype = globals().get("DTYPE", "torch.float16" if torch.cuda.is_available() and torch.cuda.get_device_capability()[0] < 8 else "torch.bfloat16")
active_small = globals().get("SMALL", "Qwen/Qwen3-0.6B")
active_instruct = globals().get("INSTRUCT", "Qwen/Qwen3-4B-Instruct-2507")
active_thinking = globals().get("THINKING", "Qwen/Qwen3-4B-Thinking-2507")
active_questions_len = len(globals().get("questions", []))
active_state_levels = list(globals().get("STATE_LEVELS", [4, 16, 64]))
active_spec_impls_keys = sorted(list(globals().get("SPEC_IMPLS", {}).keys()))

run_record = {
    "environment": {"gpu": torch.cuda.get_device_name(0) if torch.cuda.is_available() else "None", "torch": torch.__version__, "transformers": transformers.__version__,
                    "dtype_4b": str(active_dtype), "dtype_0.6b": "torch.float32"},
    "models": {"mechanism": active_small, "comparison": active_instruct, "reasoning": active_thinking},
    "generation": {"decoding": "greedy (do_sample=False)", "sampling_experiment": "temperature=0.8, top_p=0.95, n=10",
                   "answer_format_hint": globals().get("FORMAT_HINT", ""),
                   "max_new_tokens": {"direct": 24, "scratchpad_addition": 1200, "scratchpad_state": "200 + 200 * steps",
                                      "reasoning": 5000, "pal": "600 (addition) / 2500 (state, 4 and 16 steps) / 4000 (state, 64 steps)"}},
    "benchmark": {"seed": 20261011, "per_level": 5, "questions_total": active_questions_len, "state_levels_used": active_state_levels,
                  "grading": "答えの文字列と正解の十進整数の文字列の完全一致"},
    "external": {"python_subprocess_timeout_s": 10, "harness_max_rounds": 6, "tools": active_spec_impls_keys},
    "accuracy_by_condition": {c: f"{v[0]}/{v[1]}（平均 {v[2] / v[1]:.0f} トークン）" for c, v in sorted(agg.items())},
    "reasoning_experiment": ({"questions": len(e_rows), "finished": sum(r["finished"] for r in e_rows),
                              "correct": sum(r["ok"] for r in e_rows)} if "e_rows" in globals() else "未実行"),
    "not_observed": ["商用サービスの内部構成", "各層の内部計算の因果過程", "非公開の推論トークン"],
}
print(json.dumps(run_record, ensure_ascii=False, indent=2))
with open("llm_understanding_run_record.json", "w", encoding="utf-8") as f:
    json.dump(run_record, f, ensure_ascii=False, indent=2)
````

| 「答えだけ」と言っても解けるのは矛盾か | 矛盾ではない 表示を短くしても、途中の生成や非表示の推論や外部計算は残る 本当に途中を書かせないと、手順の多い問題は解けなくなる 表示だけ隠しても答えは変わらない（問12、問13） |
| 画面の思考は、中身がそのまま見えているのか | 違う 生成された推論テキスト、その要約、進捗の表示、実行記録を区別する どれも内部計算の完全な記録ではない（問14） |
| 検索や厳密な計算は、Transformer以外の仕組みか | そうである モデルが要求を書き、外側のプログラムが検索・計算を実行して結果を戻す 検索が誤れば、計算が正確でも答えは誤る（問18、問22） |
| Ollamaなどは、なぜ周辺機能まで共通に扱えるか | 共通化しているのは、モデルの実行、生成の繰り返し、入出力の形式である 能力、ツールの実装、ワークフローは共通化されない（問30〜32） |

>
> **重みの大きさ ≠ 実行した計算量 ≠ 表示された文章の長さ ≠ 正しさの保証**
